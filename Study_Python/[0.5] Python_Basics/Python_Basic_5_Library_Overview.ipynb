# 1. 시계열과 일자 계산: `datetime`

고객의 가입일, 결제일, 로그 타임스탬프를 다룰 때 필수적입니다.


```python
from datetime import datetime, timedelta

# 현재 시각 및 문자열 파싱
now = datetime.now()
print(f"현재 시각: {now.strftime('%Y-%m-%d %H:%M:%S')}")

# 로그 일자 파싱 (문자열 -> datetime 객체)
log_str = "2026-11-16 14:30:00"
log_dt = datetime.strptime(log_str, "%Y-%m-%d %H:%M:%S")

# 경과 시간 계산 (timedelta)
diff = now - log_dt
print(f"로그 발생 후 경과 시간: {diff.total_seconds():.1f}초")
```

# 2. 난수 생성과 재현성(Reproducibility): `random`

머신러닝 데이터 분할이나 A/B 테스트 그룹 배정 시 결과를 재현할 수 있도록 **랜덤 시드(Seed)**를 고정해야 합니다.


```python
import random

# 시드 고정 (동일한 난수 순서 재현)
random.seed(42)

# 1 ~ 100 사이 난수 정수 생성
rand_id = random.randint(1, 100)

# 리스트에서 비복원 표본 추출 (Sample)
user_pool = [f"User_{i}" for i in range(1, 101)]
ab_test_group = random.sample(user_pool, 5)

print(f"생성된 난수 ID: {rand_id}")
print(f"추출된 A/B 테스트 표본 5명: {ab_test_group}")
```

# 3. 데이터 과학 생태계 성능 비교: 파이썬 리스트 vs NumPy

수치 계산에서 왜 순수 파이썬 리스트 대신 NumPy와 Pandas를 써야 하는지 체감할 수 있는 벤치마크 예시입니다.
NumPy는 내부적으로 연속된 메모리 공간에 C 레벨 배열을 적재하여 수십~수백 배 빠른 고속 벡터화 연산을 제공합니다.


```python
import time
import numpy as np

# 1,000,000개 수치 데이터 생성
n_elements = 1_000_000
py_list = list(range(n_elements))
np_arr = np.arange(n_elements)

# 1. 파이썬 리스트 제곱 연산 시간 측정
start_py = time.time()
py_squared = [x ** 2 for x in py_list]
end_py = time.time()
py_duration = end_py - start_py

# 2. NumPy 배열 제곱 연산 시간 측정
start_np = time.time()
np_squared = np_arr ** 2
end_np = time.time()
np_duration = end_np - start_np

print(f"파이썬 리스트 소요 시간: {py_duration:.4f}초")
print(f"NumPy 배열 소요 시간:   {np_duration:.4f}초")
print(f"NumPy 가속 배수:         약 {py_duration / np_duration:.1f}배 빠름!")
```
