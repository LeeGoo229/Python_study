데이터는 크게 값의 의미에 따라 **범주형(Categorical)**과 **수치형(Numerical)**으로 분류됩니다.

```mermaid
flowchart TD
    Data["데이터 변수 (Variables)"] --> Cat["범주형 (Categorical)"]
    Data --> Num["수치형 (Numerical)"]
    Cat --> Nom["명목형 (Nominal)<br/>순서 없음 (성별, 지역, 혈액형)"]

```


# 1. 변수의 의미에 따른 세부 분류 체계

| 대분류 | 소분류 | 특징 및 연산 가능 여부 | 실무 데이터 예시 | 적합한 대표값 / 시각화 |
| :--- | :--- | :--- | :--- | :--- |
| **범주형** | **명목형 (Nominal)** | 순서가 없으며 단순 식별/구분 목적 | 고객 성별, 거주 도시, 결제 수단 (Card/Cash) | 최빈값, 빈도수 / 막대그래프, 파이차트 |
| **범주형** | **순서형 (Ordinal)** | 순서나 위계가 있으나 간격이 일정치 않음 | 고객 등급 (Bronze < Silver < Gold), 설문 만족도 (1~5점) | 중앙값, 최빈값 / 순서형 막대그래프 |
| **수치형** | **이산형 (Discrete)** | 셀 수 있는 정수 단위(Count)의 값 | 장바구니 담은 개수, 웹사이트 클릭 횟수 | 평균, 중앙값, 분산 / 막대그래프, 이산 히스토그램 |
| **수치형** | **연속형 (Continuous)** | 연속적인 실수 구간에서 측정되는 양 | 총 결제 금액, 체류 시간(초), 할인율 | 평균, 표준편차, 사분위수 / 히스토그램, 박스플롯, 산점도 |


# 2. 파이썬 및 Pandas 자료형(dtype)과의 대응

실제 데이터프레임에서 각 변수 유형은 다음과 같이 저장됩니다:
- **명목형/순서형:** `object` (문자열) 또는 `category` (범주형 전용 dtype)
  - `category` 자료형을 쓰면 메모리 사용량이 대폭 줄어들고, `ordered=True`를 주어 명시적인 크기 비교(`Gold > Silver`)가 가능해집니다.
- **이산형:** `int64` (또는 결측치가 있는 경우 `Int64`)
- **연속형:** `float64`


````python
import pandas as pd

# 샘플 고객 데이터셋 정의
df_customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],                   # 명목형 식별자 (int/object)
    "gender": ["F", "M", "F", "M", "F"],                         # 명목형 (Nominal)
    "membership": ["Silver", "Gold", "Bronze", "Gold", "Silver"], # 순서형 (Ordinal)
    "visit_count": [3, 12, 1, 8, 5],                             # 수치-이산형 (Discrete)
    "total_spend": [45200.5, 230000.0, 12000.0, 154000.0, 68000.0] # 수치-연속형 (Continuous)
})

print("기본 dtypes:")
print(df_customers.dtypes)

# membership 컬럼을 순서가 있는 category 타입으로 변환
grade_order = ["Bronze", "Silver", "Gold"]
df_customers["membership"] = pd.Categorical(df_customers["membership"], categories=grade_order, ordered=True)

print("\n변환 후 membership dtype:", df_customers["membership"].dtype)
print("Gold가 Silver보다 높은가?:", df_customers["membership"][1] > df_customers["membership"][0])
````

# 3. 모델링에서의 역할 구분: 독립변수(X)와 종속변수(Y)

데이터 과학 및 통계학에서 변수는 인과관계 또는 예측 모형 내 역할에 따라 다음과 같이 구분됩니다:

1. **독립변수 (Independent Variable, 설명변수, Feature, $X$):**
   - 모델의 입력값이며, 종속변수의 변화를 설명하거나 예측하는 원인 변수입니다.
   - 예: 고객의 나이, 방문 횟수, 광고 노출 시간.
2. **종속변수 (Dependent Variable, 반응변수, Target/Label, $Y$):**
   - 모델이 최종적으로 예측하고자 하는 결과 변수입니다.
   - 예: 고객의 구매 여부 (이탈 여부), 다음 달 총 구매 금액.
