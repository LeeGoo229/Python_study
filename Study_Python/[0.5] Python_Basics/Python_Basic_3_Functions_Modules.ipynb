## 처음 배우는 분을 위한 안내

- **함수(Function):** 입력값(매개변수, Parameter)을 받아 정해진 연산을 수행한 뒤 결과값(`return`)을 돌려주는 코드 묶음입니다.
- **기본 인자(Default Parameter):** 자주 쓰이는 값(예: `decimals=2`)을 기본값으로 지정하여 호출 시 생략할 수 있습니다.
- **람다식(Lambda):** 이름 없는 한 줄짜리 익명 함수(`lambda x: x * 2`)로, 데이터프레임의 `.apply()` 메서드에 즉석에서 전달할 때 유용합니다.
- **모듈(Module):** 관련된 함수들을 모아둔 파이썬 파일(`.py`)이며, `import` 키워드를 통해 다른 노트북에서 불러와 쓸 수 있습니다.


# 1. 함수 정의와 다중 반환값 (Multiple Returns)

파이썬 함수는 튜플을 활용하여 여러 개의 분석 지표를 한 번에 반환할 수 있습니다.


```python
# 요약 통계량(평균, 최솟값, 최댓값)을 한 번에 구하는 함수
def get_summary_stats(numbers):
    if not numbers:
        return 0.0, 0, 0
    avg_val = sum(numbers) / len(numbers)
    min_val = min(numbers)
    max_val = max(numbers)
    return avg_val, min_val, max_val

sales = [12000, 15000, 9800, 21000, 18500]
mean_s, min_s, max_s = get_summary_stats(sales)
print(f"평균: {mean_s:,.1f}원, 최소: {min_s:,}원, 최대: {max_s:,}원")
```

# 2. 람다(Lambda) 함수와 정렬 키(`key=lambda`)

복잡한 딕셔너리 리스트를 원하는 피처 기준으로 손쉽게 정렬할 수 있습니다.


```python
# 상품 데이터 목록 (이름, 가격, 평점)
products = [
    {"name": "Laptop", "price": 1200000, "rating": 4.8},
    {"name": "Mouse", "price": 35000, "rating": 4.5},
    {"name": "Monitor", "price": 450000, "rating": 4.9},
    {"name": "Keyboard", "price": 89000, "rating": 4.6}
]

# 평점(rating) 기준 내림차순 정렬
sorted_by_rating = sorted(products, key=lambda x: x["rating"], reverse=True)
print("=== 평점 높은 순 정렬 ===")
for p in sorted_by_rating:
    print(f"{p['name']}: {p['rating']}점 (가격: {p['price']:,}원)")
```

# 3. 고차 함수: `map`과 `filter`

리스트의 모든 원소에 변환을 가하거나 조건을 만족하는 요소만 남기는 함수형 프로그래밍 도구입니다.


```python
prices_usd = [10, 25, 40, 15]
exchange_rate = 1350

# map: 모든 원소에 환율 적용
prices_krw = list(map(lambda x: x * exchange_rate, prices_usd))
print("원화 변환 가격:", prices_krw)

# filter: 30,000원 이상인 상품만 필터링
premium_items = list(filter(lambda x: x >= 30000, prices_krw))
print("프리미엄 상품 목록:", premium_items)
```
