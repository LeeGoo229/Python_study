# 7. 제곱합 분해와 결정계수

절편을 포함해 학습한 최소제곱 회귀의 훈련 표본에서는 전체 제곱합 SST가 회귀 제곱합 SSR과 오차 제곱합 SSE의 합입니다.

원본 노트북 Lv2_Python_1119_SST_SSR_SSE__R2.ipynb

image.png

```python
# 필요한 라이브러리 불러오기
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# 예제 데이터 (x: 독립 변수, y: 종속 변수)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1.5, 1.7, 3.2, 4.0, 5.1])
```

```python
# 선형 회귀 모델 생성 및 학습
model = LinearRegression()
model.fit(X, y)
```

```python
# 예측값
y_pred = model.predict(X)
```

image.png

```python
# SST (Total Sum of Squares)
# 관측값의 전체 변동성

y_mean = np.mean(y)  # 관측값의 평균
SST = np.sum((y - y_mean) ** 2)
print(f'SST (Total Sum of Squares): {SST:.2f}')
```

image.png

```python
# SSR (Regression Sum of Squares)
# 회귀 모델이 설명한 변동성

SSR = np.sum((y_pred - y_mean) ** 2)
print(f'SSR (Regression Sum of Squares): {SSR:.2f}')
```

image.png

```python
# SSE (Sum of Squared Errors)
# 모델이 설명하지 못한 오차

SSE = np.sum((y - y_pred) ** 2)
print(f'SSE (Sum of Squared Errors): {SSE:.2f}')
```

image.png

```python
# R^2
# 회귀 모델이 데이터를 얼마나 잘 설명하는지를 나타내는 지표
R2 = model.score(X, y)  # scikit-learn의 score 메서드는 R^2 값을 반환
```

```python
print(f'R^2 (Determination Coefficient): {R2:.2f}')
```

SST: 관측값의 전체 변동성 <br>
SSR: 회귀 모델이 설명한 변동성<br>
SSE: 모델이 설명하지 못한 오차<br>
R²: 회귀 모델이 데이터를 얼마나 잘 설명하는지를 나타내는 지표

```python
# 회귀 계수 (기울기)와 절편 출력
coef = model.coef_  # 기울기 (회귀 계수)
# intercept = model.intercept_  # 절편

print(f'Regression Coefficient (Slope): {coef[0]:.2f}')
# print(f'Intercept: {intercept:.2f}')

# 예측값
y_pred = model.predict(X)
```

```python
# 데이터와 회귀선 시각화
import matplotlib.pyplot as plt

# 데이터와 회귀선, 평균선 시각화
plt.scatter(X, y, color='blue', label='Data Points')  # 실제 데이터 점
plt.plot(X, y_pred, color='red', label='Regression Line')  # 회귀선
plt.axhline(y=y_mean, color='green', linestyle='--', label='Average Line')  # 평균선


# 그래프 설정
plt.title('Linear Regression Visualization')
plt.xlabel('X (Independent Variable)')
plt.ylabel('y (Dependent Variable)')
plt.legend()

# 그래프 출력
plt.grid(True)
plt.show()
```

```python
print(model.coef_)       # 특성별 회귀계수
print(model.intercept_)  # 절편
```
