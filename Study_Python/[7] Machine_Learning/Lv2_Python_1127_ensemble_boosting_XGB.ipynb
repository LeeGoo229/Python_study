# 27. XGBoost 부스팅 분류 (XGBoost)

L1/L2 규제와 병렬 트리 구축을 지원하는 `xgboost.XGBClassifier`를 학습하고 평가 지표(`eval_metric`) 설정과 분류 정확도를 확인합니다.

```python
import time
import xgboost as xgb
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.3, random_state=42, stratify=iris.target
)

start = time.time()
xgb_clf = xgb.XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    eval_metric='mlogloss',
    random_state=42,
)
xgb_clf.fit(X_train, y_train)
elapsed = time.time() - start

print(f'XGBoost Accuracy: {accuracy_score(y_test, xgb_clf.predict(X_test)):.4f} (소요시간: {elapsed:.4f}s)')
```
