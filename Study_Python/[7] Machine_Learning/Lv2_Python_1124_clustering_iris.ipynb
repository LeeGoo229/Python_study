# 14. K-평균 군집화와 Iris 품종 비교 (K-Means Clustering)

레이블 없이 Iris 데이터의 4개 특성만으로 `KMeans` 군집화를 수행하고, 엘보우(Elbow) 곡선으로 최적 군집 수를 확인한 뒤 할당된 `Cluster_ID`와 실제 품종(`target`)을 교차표로 비교합니다.

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 내장 iris 데이터셋 로드 (특성 4개, 레이블은 비교용으로만 보관)
iris = load_iris()
X = iris.data
print('데이터 크기:', X.shape)
```

```python
# 엘보우 메소드를 사용하여 최적의 클러스터 수 탐색
inertia = []
for n_clusters in range(2, 11):  # 클러스터 개수를 2부터 10까지 늘리며 반복
    kmeans_tmp = KMeans(
        n_clusters=n_clusters,
        init='k-means++',   # 클러스터 중심(centroid)의 초기 선택 방법
        max_iter=300,       # 알고리즘의 최대 반복 횟수
        n_init=10,          # 서로 다른 무작위 중심으로 초기화 후 실행하는 횟수 (최적 결과 선택)
        random_state=0,
    )
    kmeans_tmp.fit(X)
    inertia.append(kmeans_tmp.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(range(2, 11), inertia, marker='o')
plt.title('The Elbow Method (Iris)')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# KMeans 클러스터링 실행 (n_clusters=3, k-means++ 초기화)
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, n_init=10, random_state=0)
y_kmeans = kmeans.fit_predict(X)

# 각 데이터 포인트에 할당된 클러스터 ID 및 기존 품종 정보 병합
iris_df = pd.DataFrame(X, columns=iris.feature_names)
iris_df['Cluster_ID'] = y_kmeans
iris_df['target'] = iris.target
iris_df.tail()
```

```python
# 교차표(Crosstab)로 실제 품종(target)과 군집 결과(Cluster_ID) 비교
ct = pd.crosstab(iris_df['target'], iris_df['Cluster_ID'])
print(ct)
```
