# 12. 특성 변환과 선형 분리 (Squared Features)

직선으로 나눌 수 없는 비선형 데이터를 제곱 특성 변환($x_1^2$)으로 선형 분리가 가능한 공간으로 매핑하는 원리를 시각화합니다.

```python
import matplotlib.pyplot as plt

# 임의의 예제 데이터 생성: 각 점의 좌표는 (x1, x2)
# A: (-1, 1), (1, 1) -> 파란색
# B: (-3, 1), (3, 1) -> 빨간색
x1 = [-1, 1, -3, 3]
x2 = [1, 1, 1, 1]
colors = ['blue', 'blue', 'red', 'red']
```

```python
# x1만 제곱하여 새로운 특성 공간으로 변환
x1_squared = [x ** 2 for x in x1]
print('변환 전 x1:', x1)
print('변환 후 x1^2:', x1_squared)
```

```python
# 변환 전/후 시각화 비교
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter(x1, x2, c=colors, s=100)
plt.title('Before (Linear Inseparable)')
plt.xlabel('x1')
plt.ylabel('x2')
plt.xlim(-4, 4)

plt.subplot(1, 2, 2)
plt.scatter(x1_squared, x2, c=colors, s=100)
plt.axvline(x=5, color='black', linestyle='--')
plt.title('After: x1 squared (Linear Separable)')
plt.xlabel('x1^2')
plt.xlim(0, 10)

plt.tight_layout()
plt.show()
```
