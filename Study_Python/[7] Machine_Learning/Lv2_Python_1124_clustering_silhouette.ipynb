# 16. 실루엣 스코어 군집 평가 (Silhouette Score)

군집 내부 응집도와 군집 간 분리도를 함께 측정하는 `silhouette_score`를 `k=2~10` 범위에서 계산해 최적 군집 수를 정량적으로 판별합니다.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=500, centers=4, cluster_std=0.60, random_state=0)
range_n_clusters = list(range(2, 11))
```

```python
silhouette_avg_scores = []
for n_clusters in range_n_clusters:
    clusterer = KMeans(n_clusters=n_clusters, n_init=10, random_state=10)
    cluster_labels = clusterer.fit_predict(X)
    silhouette_avg = silhouette_score(X, cluster_labels)
    silhouette_avg_scores.append(silhouette_avg)
    print(f'n_clusters = {n_clusters}, average silhouette_score = {silhouette_avg:.3f}')

plt.figure(figsize=(8, 4))
plt.plot(range_n_clusters, silhouette_avg_scores, marker='o')
plt.title('Silhouette Scores for Various Numbers of Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Average Silhouette Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
