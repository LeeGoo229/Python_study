**1. 분류 - 로지스틱 회귀**

```python
# 필요한 라이브러리 불러오기
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# 데이터 로드 (Iris 데이터셋 사용)
data = load_iris()
X = data.data        # 설명변수, 속성
y = data.target      # 목표변수, 정답, target
```

```python
# 데이터셋을 dataframe 형태로 변환
import pandas as pd

iris_df = pd.DataFrame(data=data.data, columns=data.feature_names)
iris_df['target'] = data.target
iris_df.tail()
```

```python
# target 컬럼의 각 범주별 데이터 갯수
print(iris_df['target'].value_counts())

# target 컬럼의 각 범주별 데이터 갯수 + 비율
print(iris_df['target'].value_counts(normalize=True))

################################
### 클래스 별 데이터 비율 꼭 확인
# 3개의 꽃 종 : 꽃 종별 비율 균등....
# Study ??? <-- 극심한 불균형(클래스 별 데이터 갯수) 데이터...

# 0.96 : 0.04

####### 0.95 <--- 분류 정확도  accuracy

# Base Line



# 3 : 7



################################

# target 컬럼의 각 범주별 데이터 갯수 시각화
import matplotlib.pyplot as plt
iris_df['target'].value_counts().plot(kind='bar')
plt.xlabel('Target Class')
plt.ylabel('Count')
plt.title('Number of Samples per Class')
plt.show()
```

```python
# 데이터 분할 (학습 데이터와 테스트 데이터)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
######## 분류 모델 ########
# 로지스틱 회귀 모델 생성 및 학습
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)
```

```python
# 예측 및 정확도 평가
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'분류 정확도: {accuracy * 100:.2f}%')
```

**2. 회귀 예제 (선형 회귀)**

```python
# 필요한 라이브러리 불러오기
# from sklearn.datasets import load_boston # removed
from sklearn.datasets import fetch_california_housing # Use this to load the California housing dataset
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# 데이터 로드 (California housing dataset 사용)
housing = fetch_california_housing() # Load the California housing dataset
X = housing.data
y = housing.target
```

```python
# 데이터셋을 dataframe 형태로 변환
import pandas as pd

housing_df = pd.DataFrame(data=housing.data, columns=housing.feature_names)
housing_df['target'] = housing.target
housing_df.head()
```

```python
# MedInc : 가구의 중간 소득 (10,000 달러 단위)
# HouseAge : 주택의 나이 (년)
# AveRooms : 평균 방 개수
# AveBedrms : 평균 침실 개수
# Population : 블록 내 인구 수
# AveOccup : 평균 가구원 수
# Latitude : 위도
# Longitude : 경도
```

```python
# 데이터 분할 (학습 데이터와 테스트 데이터)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
######## 회귀 모델 ########
# 선형 회귀 모델 생성 및 학습
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# 예측 및 MSE(평균 제곱 오차) 평가
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
print(f'회귀 MSE: {mse:.2f}')
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error: {mae}")
### 4.526  4억 5천
####           5천
```
