# 22. 에이다부스트 앙상블 분류 (AdaBoost)

오분류된 샘플의 가중치를 단계적으로 높여 약한 학습기를 결합하는 `AdaBoostClassifier`를 학습하고 기반 트리 깊이(`max_depth`)에 따른 성능 변화를 비교합니다.

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 분류 데이터 생성 및 훈련/테스트 분할
X, y = make_classification(n_samples=1000, n_features=20, n_informative=2, n_redundant=0, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 기본 AdaBoost 모델 (기본 estimator: DecisionTreeClassifier(max_depth=1))
ada_default = AdaBoostClassifier(n_estimators=100, random_state=42)
ada_default.fit(X_train, y_train)
print('기본 기반 모델:', ada_default.estimator)
print(f'AdaBoost (max_depth=1) Accuracy: {accuracy_score(y_test, ada_default.predict(X_test)):.4f}')
```

```python
# 기반 트리의 깊이를 max_depth=3으로 조정한 AdaBoost 비교
ada_deep = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=3, random_state=42),
    n_estimators=100,
    learning_rate=0.5,
    random_state=42,
)
ada_deep.fit(X_train, y_train)
print(f'AdaBoost (max_depth=3, lr=0.5) Accuracy: {accuracy_score(y_test, ada_deep.predict(X_test)):.4f}')
```
