# 26. LightGBM 부스팅 분류 (LightGBM)

리프 중심 트리 분할(Leaf-wise)로 빠른 학습 속도와 높은 효율을 제공하는 `lightgbm.LGBMClassifier`를 Iris와 합성 분류 데이터에 적용해 학습 시간과 정확도를 확인합니다.

```python
import time
import warnings
import lightgbm as lgb
from sklearn.datasets import load_iris, make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

warnings.filterwarnings('ignore')

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.3, random_state=42, stratify=iris.target
)

start = time.time()
lgb_clf = lgb.LGBMClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42, verbosity=-1)
lgb_clf.fit(X_train, y_train)
elapsed = time.time() - start

print(f'Iris LightGBM Accuracy: {accuracy_score(y_test, lgb_clf.predict(X_test)):.4f} (소요시간: {elapsed:.4f}s)')
```

```python
# 합성 분류 데이터(1000개 샘플, 20개 특성)에서 LightGBM 평가
X2, y2 = make_classification(n_samples=1000, n_features=20, n_informative=2, n_redundant=0, random_state=42)
X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size=0.3, random_state=42)

lgb_clf2 = lgb.LGBMClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42, verbosity=-1)
lgb_clf2.fit(X_train2, y_train2)
print(f'Synthetic LightGBM Accuracy: {accuracy_score(y_test2, lgb_clf2.predict(X_test2)):.4f}')
```
