# 3. 엔트로피와 특성 중요도

분류 트리는 엔트로피를 줄이는 분할을 찾습니다. 불순도 기반 중요도는 해당 모델이 사용한 신호를 요약합니다.

원본 노트북 Lv2_Python_1118_FeatureSelection_Entropy.ipynb

**Feature selection - Entropy**

```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

from sklearn.model_selection import train_test_split
```

```python
# Iris 데이터셋 로드
iris = load_iris()
X = iris.data      #  설명변수
y = iris.target    #  종속변수, Target, 정답,
```

```python
df = pd.DataFrame(iris.data)
df.head()
```

```python
# 학습 데이터와 테스트 데이터로 분리
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# 의사결정 나무 모델 생성 (분할 기준으로 엔트로피(정보이득) 사용)
model = DecisionTreeClassifier(criterion='entropy', random_state=42)  # criterion='entropy' : 분할 기준으로 엔트로피(정보이득) 사용
model.fit(X_train, y_train)

# , random_state=42
```

```python
# 각 특성의 중요도 출력
#  중요도 <- 엔트로피를 기준으로 계산된 각 특성의 중요도
#  값이 높을수록 해당 특성이 모델의 결정에 중요한 역할을 한다는 의미

importances = model.feature_importances_
for i, importance in enumerate(importances):
    print(f"특성 {i}: 중요도 {importance:.4f}")
         # 특성 번호(i) 중요도(importance) 출력
         # .4f 소수점 아래 넷째 자리 반올림

# model.feature_importances_ : 학습된 model 내부에 저장된 특성 중요도(배열)
  # 각 특성이 데이터를 나누는 데(불순도 감소에) 기여한 정도
# enumerate() 함수 : importances 배열 안의 중요도 값을 하나씩 꺼내면서,순서 번호(인덱스) i를 0부터 자동으로 생성
```

```python
df = pd.DataFrame(X, columns=iris.feature_names)

# 각 특성의 중요도 확인
imp = pd.Series(model.feature_importances_, index=iris.feature_names).sort_values(ascending=False)
print(imp)

#  중요도 <- 엔트로피를 기준으로 계산된 각 특성의 중요도
#  값이 높을수록 해당 특성이 모델의 결정에 중요한 역할을 한다는 의미

#### TASK ... 시각화 결과와 비교
```

```python
# 모델 정확도 평가
accuracy = model.score(X_test, y_test)
print(f"모델 정확도: {accuracy:.4f}")
```
