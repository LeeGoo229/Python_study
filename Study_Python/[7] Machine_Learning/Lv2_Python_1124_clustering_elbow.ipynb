# 15. 엘보우 기법으로 최적 군집 수 찾기 (Elbow Method)

`make_blobs`로 4개 중심을 가진 가상 데이터를 생성하고, 군집 수(`k=2~10`)별 이너셔(`inertia_`) 변화를 시각화해 엘보우 지점을 탐색합니다.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 가상의 군집(Cluster) 데이터 생성 (n_samples는 300 유지)
# n_samples=300: 300개 샘플 / centers=4: 4개 클러스터 중심 / cluster_std=0.60: 각 군집의 퍼짐 정도
# 언더스코어(_)는 반환된 실제 레이블을 군집 학습에 사용하지 않기 위해 무시
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# 2차원 산점도 시각화
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], s=15, alpha=0.6)
plt.title('2D Scatter Plot of Generated Blobs')
plt.xlabel('Feature 1 (X)')
plt.ylabel('Feature 2 (Y)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# 엘보우 메소드: k=2부터 10까지 이너셔(Inertia) 계산
# 이너셔(inertia_): 각 데이터 포인트와 소속 군집 중심점(Centroid) 사이의 거리 제곱의 합 (응집도)
# 군집 개수가 증가할수록 이너셔는 줄어들며, 급격히 줄다가 완만해지는 꺾이는 지점(팔꿈치)을 최적 k로 선택
inertia = []
for n_clusters in range(2, 11):
    kmeans = KMeans(n_clusters=n_clusters, init='k-means++', max_iter=300, n_init=10, random_state=0)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)
    print(f'k={n_clusters}: inertia={kmeans.inertia_:.2f}')

plt.figure(figsize=(7, 4))
plt.plot(range(2, 11), inertia, marker='o')
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
