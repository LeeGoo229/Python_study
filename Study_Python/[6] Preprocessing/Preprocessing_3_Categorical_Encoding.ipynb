# 전처리 3 · 범주형 데이터 인코딩 (One-Hot, Label, Ordinal) — 개념 정리

대부분의 통계 머신러닝 모델(선형 회귀, 로지스틱 회귀, 신경망, SVM 등)은 문자열 텍스트를 직접 입력받을 수 없으며, 오직 **수치형 행렬(Numerical Matrix)**만을 계산할 수 있습니다.
따라서 성별, 지역, 상품 카테고리 등의 범주형 피처를 수학적으로 올바르게 숫자로 변환하는 **인코딩(Encoding)** 작업이 필수적입니다.

이 단원에서는 `pd.get_dummies()`, 사이킷런의 `OneHotEncoder`, 타겟 레이블 변환을 위한 `LabelEncoder`, 순서형 변수를 위한 `OrdinalEncoder`, 그리고 희소 범주(Rare Categories) 병합 기법을 학습합니다.
실습 문제는 `Preprocessing_3_Categorical_Encoding_Practice.ipynb`, 해답은 `Preprocessing_3_Categorical_Encoding_Answer.ipynb`에 수록되어 있습니다.


## 주요 인코딩 기법 비교

| 인코딩 방식 | 변환 메커니즘 | 적합한 변수 유형 | 장점 및 주의점 |
| :--- | :--- | :--- | :--- |
| **원-핫 인코딩 (One-Hot)** | 각 범주마다 0과 1을 갖는 독립 더미 컬럼 생성 | 명목형 (순서 없음) | 인위적 서열을 부여하지 않음. 단, 범주 수가 많으면 차원 폭증 |
| **레이블 인코딩 (Label)** | 범주를 0, 1, 2, 3 정수로 1:1 매핑 | 종속변수 타겟($Y$) | 독립변수에 쓰면 모델이 숫자 크기(2 > 1)를 가중치로 오인 |
| **순서형 인코딩 (Ordinal)** | 명시적 서열 순서에 따라 정수 할당 | 순서형 (등급, 만족도) | 순서 정보 보존 (Bronze=0, Silver=1, Gold=2) |
