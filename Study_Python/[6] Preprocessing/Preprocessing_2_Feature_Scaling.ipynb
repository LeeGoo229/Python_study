# 전처리 2 · 피처 스케일링(Feature Scaling)과 데이터 누수 방지 — 개념 정리

서로 다른 측정 단위와 스케일을 갖는 수치 피처들(예: 나이 20~60 vs 연봉 3000만~1억)을 그대로 머신러닝 모형에 투입하면, 단위가 큰 피처가 거리 계산이나 경사 하강법 학습을 압도하여 심각한 왜곡이 발생합니다.
특히 거리 기반 알고리즘(KNN, K-Means, SVM, PCA)이나 정규화 선형 회귀(Ridge, Lasso), 딥러닝 신경망에서는 **피처 스케일링이 필수 선행 조건**입니다.

이 단원에서는 `StandardScaler`, `MinMaxScaler`, `RobustScaler`, `Normalizer` 4대 스케일러의 수학적 원리를 비교하고, 실무에서 가장 흔히 범하는 **데이터 누수(Data Leakage)** 방지 원칙을 마스터합니다.
실습 문제는 `Preprocessing_2_Feature_Scaling_Practice.ipynb`, 해답은 `Preprocessing_2_Feature_Scaling_Answer.ipynb`에 수록되어 있습니다.


## 처음 배우는 분을 위한 안내: 4대 스케일러 비교

| 스케일러 | 수학 공식 | 변환 후 스케일 | 주요 특징 및 적합한 상황 |
| :--- | :--- | :--- | :--- |
| **StandardScaler** | $z = \frac{x - \mu}{\sigma}$ | 평균 0, 표준편차 1 | 정규분포를 따르는 피처, 선형 회귀, 로지스틱 회귀, PCA |
| **MinMaxScaler** | $x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}$ | [0, 1] 고정 구간 | 이미지 픽셀(0~255), 활성화 함수(Sigmoid/ReLU)가 있는 신경망 |
| **RobustScaler** | $x_{scaled} = \frac{x - Q2}{Q3 - Q1}$ | 중앙값 0, IQR 1 | 이상치가 포함된 피처 (중앙값과 사분위수를 써서 이상치에 강건) |
| **Normalizer** | $x_{norm} = \frac{x}{\|x\|_2}$ | 각 행의 벡터 길이 = 1 | 텍스트 단어 빈도(TF-IDF), 코사인 유사도, 추천 시스템 |

> [!CAUTION] 데이터 누수(Data Leakage) 엄금 규칙!
> 스케일러의 기준값($\mu, \sigma, x_{min}, x_{max}$)을 학습할 때 반드시 **훈련용 데이터(Train Set)만을 사용해 `.fit()`**해야 합니다.
> 시험용 데이터(Test Set)는 훈련셋에서 학습된 기준값으로 **`.transform()`만 수행**해야 합니다. 테스트셋에 `.fit()`을 호출하면 미래 정보가 유출되어 실전 배포 시 모델 성능이 폭락합니다.
