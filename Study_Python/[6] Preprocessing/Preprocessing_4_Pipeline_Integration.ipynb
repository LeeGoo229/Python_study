# 전처리 4 · ColumnTransformer와 머신러닝 파이프라인(Pipeline) 통합 — 개념 정리

현실의 정형 데이터셋은 수치형 피처(나이, 소득)와 범주형 피처(도시, 기기)가 섞여 있어, 각 피처 유형마다 서로 다른 전처리 기법을 적용해야 합니다.
이를 수작업으로 나누어 전처리하면 코드가 복잡해지고 교차 검증 시 데이터 누수가 발생하기 쉽습니다.

사이킷런의 **`ColumnTransformer`**와 **`Pipeline`**은:
1. 수치형 피처: 결측치 중앙값 대체(`SimpleImputer`) + 표준화(`StandardScaler`)
2. 범주형 피처: 결측치 최빈값 대체(`SimpleImputer`) + 원-핫 인코딩(`OneHotEncoder`)
3. 최종 예측 모형: 분류기(`LogisticRegression`) 결합
위 모든 단계를 단 하나의 통합 객체로 캡슐화하여 단 한 번의 `.fit()`과 `.predict()`로 엔드투엔드 서빙을 가능하게 합니다.

실습 문제는 `Preprocessing_4_Pipeline_Integration_Practice.ipynb`, 해답은 `Preprocessing_4_Pipeline_Integration_Answer.ipynb`에 수록되어 있습니다.


## 엔드투엔드 파이프라인 아키텍처

```mermaid
flowchart TD
    Raw["원천 데이터 (X)"] --> Split{"피처 유형 자동 분기"}
    Split -->|"수치형 피처"| NumImpute["SimpleImputer (strategy='median')"]
    NumImpute --> NumScale["StandardScaler()"]
    Split -->|"범주형 피처"| CatImpute["SimpleImputer (strategy='most_frequent')"]
    CatImpute --> CatEncode["OneHotEncoder(handle_unknown='ignore')"]
    NumScale --> Concat["ColumnTransformer 피처 병합"]
    CatEncode --> Concat
    Concat --> Model["머신러닝 모델 (LogisticRegression)"]
    Model --> Pred["최종 예측 (y_pred)"]
```
