# 전처리 1 · 결측치(Missing Value)와 이상치(Outlier) 정제 — 개념 정리

현실의 원천 데이터(Raw Data)는 시스템 오류, 사용자의 입력 누락, 통신 지연 등으로 인해 무수한 결측치와 비상식적인 극단값(Outlier)을 포함하고 있습니다.
"Garbage In, Garbage Out(쓰레기가 들어가면 쓰레기가 나온다)"이라는 데이터 과학의 격언처럼, 전처리가 부실한 데이터로 학습된 머신러닝 모델은 실무에서 치명적인 오작동을 초래합니다.

이 단원에서는 결측 패턴의 진단, 대푯값 대체(평균/중앙값/최빈값), 시계열 전후방 대체(`ffill`/`bfill`), 그리고 IQR 기반 이상치 윈저화(Winsorization) 클리핑을 실습합니다.
실습 문제는 `Preprocessing_1_Missing_Outliers_Practice.ipynb`, 해답은 `Preprocessing_1_Missing_Outliers_Answer.ipynb`에 수록되어 있습니다.


## 처음 배우는 분을 위한 안내

- **결측치(Missing Value):** 데이터가 비어 있는 상태(`np.nan`, `None`).
  - **삭제(Drop):** 결측 행/열의 비중이 매우 낮고(5% 미만) 무작위로 발생했을 때 사용.
  - **대체(Imputation):** 결측 비중이 높거나 표본 손실이 클 때 평균, 중앙값, 최빈값, 또는 머신러닝 예측값으로 채움.
- **이상치(Outlier):** 대다수 관측치와 현저히 동떨어진 극단값.
  - 단순 삭제보다 상/하한선으로 눌러주는 **클리핑(Clipping / Winsorizing)**이 데이터 손실을 방지하는 안전한 실무 기법입니다.
