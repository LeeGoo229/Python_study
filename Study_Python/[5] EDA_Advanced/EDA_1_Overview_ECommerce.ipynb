# 1. 이상치 탐지의 정석: IQR (Interquartile Range) 규칙

사분위수 범위($IQR = Q3 - Q1$)를 활용하여 상한/하한 경계값을 설정합니다:
- **정상 상한:** $Q3 + 1.5 \times IQR$
- **정상 하한:** $Q1 - 1.5 \times IQR$


```python
import pandas as pd
import numpy as np

# 이커머스 샘플 데이터
df_sales = pd.DataFrame({
    "order_id": range(101, 111),
    "category": ["Tech", "Book", "Food", "Tech", "Book", "Tech", "Food", "Book", "Tech", "Food"],
    "revenue": [120000, 25000, 18000, 150000, 32000, 2200000, 15000, 41000, 135000, 21000], # 220만 원 극단값
    "delivery_days": [2, 1, 3, 2, 1, 7, 2, 1, 3, 2],
    "rating": [4.8, 4.5, 4.2, 4.9, 4.0, 2.1, 4.6, 4.3, 4.7, 4.1]
})

# revenue IQR 계산
q1 = df_sales["revenue"].quantile(0.25)
q3 = df_sales["revenue"].quantile(0.75)
iqr = q3 - q1
upper_bound = q3 + 1.5 * iqr

outliers = df_sales[df_sales["revenue"] > upper_bound]
print(f"Q1: {q1:,}원, Q3: {q3:,}원, IQR: {iqr:,}원, 상한: {upper_bound:,}원")
print("이상치 관측치:\n", outliers[["order_id", "category", "revenue"]])
```
