헬스케어 및 의료 임상 데이터(Healthcare Clinical Analytics)는 데이터의 오류가 환자의 생명과 직결되므로 가장 엄격한 탐색적 분석이 요구되는 분야입니다.

실습 문제는 `EDA_2_Healthcare_Heart_Disease_Practice.ipynb`, 모범 해답은 `EDA_2_Healthcare_Heart_Disease_Answer.ipynb`에 수록되어 있습니다.


## 주요 임상 피처 정의 (Data Dictionary)

| 컬럼명 | 의학적 의미 | 단위 및 척도 | 정상 범위 / 범주 |
| :--- | :--- | :--- | :--- |
| `age` | 환자 연령 | 세 (연속형) | 성인 29 ~ 77세 |
| `sex` | 생물학적 성별 | 범주형 | 1 = 남성, 0 = 여성 |
| `cp` | 흉통 유형 (Chest Pain) | 서열형 (0~3) | 0: 전형적 협심증, 1: 비전형적, 2: 비협심증, 3: 무증상 |
| `trestbps` | 안정 시 혈압 (Resting Blood Pressure) | mm Hg (연속형) | 정상: 120 미만, 고혈압: 140 이상 |
| `chol` | 혈청 콜레스테롤 (Cholesterol) | mg/dl (연속형) | 정상: 200 미만, 경계: 200~239, 위험: 240 이상 |
| `thalach` | 달성된 최대 심박수 (Max Heart Rate) | bpm (연속형) | 연령별 상이 (통상 70 ~ 200) |
| `target` | 심장 질환 진단 여부 | 이진 범주형 | 1 = 질환 유(Heart Disease), 0 = 정상 |
