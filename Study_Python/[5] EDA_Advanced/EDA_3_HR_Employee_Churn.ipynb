# EDA 심화 3 · HR 인사 데이터 결측 분석과 직원 이직 요인 탐색 — 개념 정리

인사(HR, Human Resources) 데이터 분석은 기업의 핵심 인재 유출을 사전에 방어하고 교육 프로그램의 효과성을 검증하는 데 핵심적인 역할을 합니다.
특히 HR 데이터는 직원들의 경력 연수, 이전 직장 이직 주기, 회사 규모 등에서 **비정형 결측치(Missing Values)**가 대량으로 발생하는 특성이 있습니다.

이 단원에서는 `hr_data.csv`를 기반으로 결측 패턴 파악, 도시 개발 지수(CDI), 교육 시간과 이직 의향(`target`) 간의 인과관계를 탐색합니다.
실습 문제는 `EDA_3_HR_Employee_Churn_Practice.ipynb`, 모범 해답은 `EDA_3_HR_Employee_Churn_Answer.ipynb`에서 확인할 수 있습니다.


## 주요 HR 피처 정의

- `city_development_index` (CDI): 거주 도시의 경제·사회 발전 지수 (0 ~ 1, 연속형)
- `relevent_experience`: 데이터/직무 관련 실무 경험 유무 (`Has relevent experience` / `No relevent experience`)
- `education_level`: 최종 학력 (`Graduate`, `Masters`, `Phd`)
- `experience`: 총 경력 연수 (`<1`, `1`~`20`, `>20`)
- `company_size`: 재직 중인 회사 규모 (직원 수)
- `training_hours`: 이수한 총 교육 훈련 시간 (시간 단위)
- `target`: 이직 희망 여부 (1 = 이직 준비 중/퇴사 의향 있음, 0 = 현 직장 잔류)


# 1. 결측치 유형과 시각적 탐색

결측치는 단순히 `dropna()`로 지우는 것이 아니라, **완전 무작위 결측(MCAR)**인지 특정 직무/조건에 결측이 몰린 **조건부 결측(MAR)**인지를 식별해야 편향을 막을 수 있습니다.
