# EDA 심화 4 · 다변량 분석(Multivariate), 교차표와 상관계수 히트맵 — 개념 정리

단일 변수(Univariate)나 두 변수 간 관계(Bivariate)를 넘어 3개 이상의 변수가 상호작용하는 복합 다차원 데이터를 분석하는 **다변량 탐색(Multivariate Analysis)** 기법을 다룹니다.
현실 세계의 비즈니스 현상은 한 가지 요인으로만 설명되지 않으며, 여러 변수를 동시에 교차 통제했을 때 비로소 숨겨진 진실(예: 심슨의 역설, 교락 효과)이 드러납니다.

이 단원에서는 다차원 피벗 테이블, 정규화 교차표(`crosstab`), 상관계수 행렬 및 다변량 이상치 탐지 기법을 체계적으로 훈련합니다.
실습 문제는 `EDA_4_Multivariate_Correlation_Practice.ipynb`, 모범 해답은 `EDA_4_Multivariate_Correlation_Answer.ipynb`에 수록되어 있습니다.


## 처음 배우는 분을 위한 안내

- **교차표 (`pd.crosstab`):** 두 개 이상의 범주형 변수 간 결합 빈도와 조건부 확률을 산출합니다. `margins=True`로 총계를 표시하고 `normalize='index'`로 행 비율을 계산합니다.
- **다차원 피벗 테이블 (`pivot_table`):** 행(Row), 열(Column), 값(Value), 집계함수(aggfunc)를 자유롭게 조합하여 3차원 이상의 지표를 요약합니다.
- **다중공선성(Multicollinearity):** 독립변수들끼리 강한 상관관계($r > 0.8$)를 보일 경우 머신러닝 회귀 모델의 계수가 왜곡될 수 있으므로 EDA 단계에서 상관 행렬을 통해 사전에 걸러내야 합니다.
