NumPy의 가장 강력한 핵심 무기는 바로 **벡터화 연산(Vectorized Operations)**과 **브로드캐스팅(Broadcasting)**입니다.
파이썬의 느린 `for` 반복문을 완전히 걷어내고, CPU의 SIMD(Single Instruction Multiple Data) 병렬 처리 기능을 활용하여 수백만 건의 수치 행렬을 눈 깜짝할 사이에 계산합니다.

## 처음 배우는 분을 위한 안내

- **벡터화(Vectorization):** 배열의 모든 요소에 대해 한 번의 표현식(`a + b`, `arr * 2`)으로 전체 원소 연산을 일괄 수행하는 메커니즘입니다.
- **브로드캐스팅(Broadcasting):** 차원이나 형상이 서로 다른 두 배열을 연산할 때, NumPy가 작은 쪽의 배열을 가상으로 늘려(복사하지 않고 메모리 효율적으로) 형상을 일치시키는 마법 같은 기능입니다.
- **축(Axis)의 이해:**
  - 2차원 행렬에서 `axis=0`은 **열(Column) 기준 연산**(위에서 아래로 압축하여 각 열의 요약치 반환)
  - `axis=1`은 **행(Row) 기준 연산**(좌에서 우로 압축하여 각 행의 요약치 반환)


# 1. 요소별 사칙연산과 유니버설 함수 (ufunc)


```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

# 요소별(Element-wise) 기본 사칙연산
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)

# 유니버설 수학 함수 (ufunc)
print("제곱근 (np.sqrt):", np.sqrt(a))
print("자연로그 (np.log):", np.log(a))
```

# 2. 브로드캐스팅 (Broadcasting)의 원리

스칼라와 배열의 연산, 혹은 1차원 벡터와 2차원 행렬 간의 연산에서 일어납니다.


```python
# 3행 3열 행렬
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# 1. 스칼라 브로드캐스팅: 모든 원소에 +5 적용
print("스칼라 덧셈:\n", matrix + 5)

# 2. 1차원 벡터 브로드캐스팅: 각 열마다 [1, 2, 3] 가산
col_bias = np.array([1, 2, 3])
print("열 기준 벡터 덧셈:\n", matrix + col_bias)
```

# 3. 축(Axis) 기준 집계 연산

데이터 분석에서 피처별(열별) 평균을 낼 것인가, 샘플별(행별) 합계를 낼 것인가를 결정짓는 핵심 파라미터입니다.


```python
scores = np.array([
    [85, 90, 78], # 학생 1의 국/영/수
    [92, 88, 95], # 학생 2의 국/영/수
    [70, 75, 80], # 학생 3의 국/영/수
    [60, 65, 70]  # 학생 4의 국/영/수
])

# axis=0: 각 과목별(열별) 평균 점수
subject_mean = scores.mean(axis=0)
print("과목별 평균 (국, 영, 수):", subject_mean)

# axis=1: 각 학생별(행별) 총점
student_sum = scores.sum(axis=1)
print("학생별 총점:", student_sum)
```

# 4. 조건 분기(`np.where`)와 행렬 곱(`@`)


```python
# np.where(조건, 참일때값, 거짓일때값)
readings = np.array([12, -5, 23, 0, -8, 15])
# 음수는 0으로 보정하고 양수는 그대로 유지
clipped = np.where(readings < 0, 0, readings)
print("음수 보정 결과:", clipped)

# 행렬 곱 (@ 또는 np.dot)
X = np.array([[1, 2], [3, 4]])
W = np.array([[0.5], [1.5]])
y_pred = X @ W
print("선형 회귀 예측값 (X @ W):\n", y_pred)
```
