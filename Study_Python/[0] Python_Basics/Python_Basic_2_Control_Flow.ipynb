## 처음 배우는 분을 위한 안내

- **조건문 (`if-elif-else`):** 참/거짓 조건에 따라 실행할 코드 블록을 나눕니다. 논리 연산자(`and`, `or`, `not`)를 조합하여 복합 조건을 평가합니다.
- **반복문 (`for`, `while`):** 리스트나 시퀀스를 하나씩 꺼내어 순회합니다. 인덱스가 필요할 땐 `enumerate()`, 여러 리스트를 동시에 묶을 땐 `zip()`을 씁니다.
- **리스트 컴프리헨션(List Comprehension):** `[x for x in data if condition]` 형태로 기존 for 루프 대비 간결하고 실행 속도가 빠른 파이써닉(Pythonic)한 문법입니다.
- **예외 처리 (`try-except`):** 숫자로 변환할 수 없는 결측치 문자열(`"N/A"`, `"-"`) 등을 안전하게 우회 처리합니다.


# 1. 조건문과 논리 연산자

비즈니스 룰에 따라 고객 등급을 판정하거나 이상 거래를 탐지할 때 사용합니다.


```python
# 조건문 기본 구조
attendance_rate = 85.0   # 출석률 (%)
assignment_count = 4     # 과제 제출 횟수

# 수료 기준: 출석률 80% 이상 AND 과제 3회 이상
if attendance_rate >= 80.0 and assignment_count >= 3:
    status = "수료 완료 (PASS)"
elif attendance_rate >= 80.0 or assignment_count >= 3:
    status = "보충 대상 (CONDITIONAL)"
else:
    status = "수료 불가 (FAIL)"

print(f"판정 결과: {status}")
```

# 2. 반복문과 파이썬 유틸리티 (`enumerate`, `zip`)

데이터 목록을 순회할 때 단순 range 대신 내장 함수를 쓰면 코드가 훨씬 깔끔해집니다.


```python
# enumerate: 인덱스와 요소를 동시에 튜플로 반환
channels = ["Search", "Facebook", "Email", "Direct"]
for rank, ch in enumerate(channels, start=1):
    print(f"{rank}위 유입 채널: {ch}")

# zip: 길이가 같은 복수의 리스트를 1:1로 묶음
users = [101, 102, 103]
revenues = [15000, 48000, 92000]
for u, rev in zip(users, revenues):
    print(f"고객 ID: {u} -> 발생 매출: {rev:,}원")
```

# 3. 리스트 컴프리헨션 (List Comprehension)

데이터 분석가가 필터링과 전처리를 단 한 줄로 끝낼 수 있는 핵심 기술입니다.


```python
# 기존 for 루프 방식
raw_scores = [85, -1, 92, 0, 78, -99, 95]
valid_scores = []
for s in raw_scores:
    if s > 0:
        valid_scores.append(s)

# 리스트 컴프리헨션 방식 (빠르고 간결함)
clean_scores = [s for s in raw_scores if s > 0]
print("정제된 점수 리스트:", clean_scores)
```

# 4. 예외 처리 (Exception Handling: `try - except`)

대량의 CSV 파일을 읽다 보면 일부 행에 결측 문자열이 섞여 `int()` 변환 시 에러가 발생합니다.
이때 `try-except` 구문으로 오류를 격리합니다.


```python
raw_data_stream = ["1200", "3400", "N/A", "5600", "error_code", "7800"]
cleaned_numbers = []

for item in raw_data_stream:
    try:
        val = int(item)
        cleaned_numbers.append(val)
    except ValueError:
        print(f"경고: 유효하지 않은 숫자 '{item}' 건너뜀")

print("안전하게 추출된 수치 목록:", cleaned_numbers)
```
