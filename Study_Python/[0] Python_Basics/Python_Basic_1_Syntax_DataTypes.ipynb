## 처음 배우는 분을 위한 안내

- **변수(Variable):** 데이터를 메모리에 기억해두는 '이름표'입니다. 파이썬은 변수에 값을 대입할 때 자료형이 자동으로 결정되는 동적 타이핑(Dynamic Typing) 언어입니다.
- **불변(Immutable) vs 가변(Mutable):** 숫자(`int`, `float`), 문자열(`str`), 튜플(`tuple`)은 한 번 만들어지면 수정할 수 없는 불변 객체입니다. 반면 리스트(`list`), 딕셔너리(`dict`), 세트(`set`)는 내용을 자유롭게 변경할 수 있는 가변 객체입니다.
- **Pandas 연계:** 파이썬의 `list`는 Pandas `Series`의 기반이 되고, `dict`는 테이블 형태의 `DataFrame`을 생성하는 핵심 재료가 됩니다.


# 1. 스칼라 자료형 (Scalar Types)

하나의 단일 값을 저장하는 가장 기본적인 자료형들입니다.

| 파이썬 자료형 | 설명 | 실무 데이터 예시 | Pandas 대응 dtype |
| :--- | :--- | :--- | :--- |
| `int` | 정수형 (크기 제한 없음) | 회원 ID, 방문 횟수, 재고 수량 | `int64` |
| `float` | 부동소수점 (실수형) | 매출액, 할인율, 전환율 (3.14, 0.05) | `float64` |
| `bool` | 참/거짓 논리형 (`True`, `False`) | 구매 여부, 활성 회원 플래그 | `bool` |
| `str` | 문자열 (텍스트 데이터) | 상품명, 고객 이름, 카테고리 코드 | `object`, `string` |


```python
# 변수 할당과 type() 함수로 자료형 확인
customer_id = 1004          # int
avg_order_value = 45200.75   # float
is_vip = True               # bool
category = "Electronics"    # str

print(f"고객 ID: {customer_id} (타입: {type(customer_id)})")
print(f"객단가: {avg_order_value} (타입: {type(avg_order_value)})")
print(f"VIP 여부: {is_vip} (타입: {type(is_vip)})")
print(f"카테고리: {category} (타입: {type(category)})")
```

> [!NOTE] 부동소수점 오차(Floating Point Precision)
> 컴퓨터는 2진수로 숫자를 표현하므로 `0.1 + 0.2` 연산 결과는 정확히 `0.3`이 아니라 `0.30000000000000004`가 됩니다.
> 금융 및 지표 분석에서는 `round(val, 2)`나 `math.isclose()` 함수를 활용하여 반올림 및 오차 범위를 보정해야 합니다.


# 2. 문자열(String) 다루기와 f-string 포맷팅

데이터 전처리에서 고객명, 주소, 비정형 텍스트를 파싱할 때 문자열 인덱싱과 슬라이싱, 포맷팅이 빈번하게 사용됩니다.


```python
# 문자열 슬라이싱 [시작:끝:간격]
order_code = "ORD-20261116-98765"

prefix = order_code[:3]       # "ORD"
order_date = order_code[4:12] # "20261116"
serial_no = order_code[13:]   # "98765"

print(f"주문 구분: {prefix}, 일자: {order_date}, 일련번호: {serial_no}")

# 문자열 정제 메서드
raw_text = "  [Special Promotion] iPhone 16 Pro Max!  "
cleaned = raw_text.strip().lower().replace("iphone", "apple_phone")
print(f"정제 전: '{raw_text}'")
print(f"정제 후: '{cleaned}'")
```

# 3. 컬렉션 자료형 (Collection Types)

여러 개의 데이터를 한 데 묶어 관리하는 자료구조입니다.

1. **리스트 (`list`):** 순서(Index)가 있고 수정 가능한 배열.
2. **튜플 (`tuple`):** 순서가 있지만 값을 수정할 수 없어 변경되면 안 되는 설정값/좌표에 사용.
3. **딕셔너리 (`dict`):** Key-Value 쌍으로 저장되어 빠른 검색(`O(1)`)이 가능한 구조.
4. **세트 (`set`):** 중복을 허용하지 않고 순서가 없는 집합 구조.


```python
# 리스트: 순서가 있고 가변적인 데이터 목록
monthly_sales = [1200, 1450, 1380, 1920]
monthly_sales.append(2100) # 5월 매출 추가
monthly_sales[0] = 1250    # 1월 매출 수정
print("월별 매출 리스트:", monthly_sales)

# 튜플: 변경 불가능한 메타데이터
image_resolution = (1920, 1080)

# 딕셔너리: 고객 프로필 및 메타데이터
customer = {
    "name": "홍길동",
    "age": 32,
    "grade": "GOLD",
    "interests": ["Data Science", "Python"]
}
print(f"고객 등급: {customer.get('grade')}")
customer["total_spend"] = 1540000 # 새로운 피처 추가

# 세트: 고유값(Unique) 추출 및 집합 연산
user_actions_day1 = {"click", "view", "cart"}
user_actions_day2 = {"view", "purchase", "review"}
common_actions = user_actions_day1 & user_actions_day2 # 교집합
all_actions = user_actions_day1 | user_actions_day2    # 합집합
print("공통 행동(교집합):", common_actions)
print("전체 행동(합집합):", all_actions)
```
