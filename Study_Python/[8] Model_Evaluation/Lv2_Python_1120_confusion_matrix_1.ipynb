# 1. 혼동 행렬 기초와 시각화 (Confusion Matrix 1)

이진 분류의 실제값(`y_true`)과 예측값(`y_pred`)으로 `confusion_matrix`를 계산하고, TN·FP·FN·TP의 위치와 `seaborn.heatmap` 시각화 방법을 학습합니다.

원본 노트북 Lv2_Python_1120_confusion_matrix_1.ipynb

image.png

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 실제 값과 예측 값
y_true = [1, 0, 0, 1, 1, 0, 1, 0]
y_pred = [1, 1, 0, 1, 0, 0, 1, 0]
```

```python
# 혼동 행렬 계산
conf_matrix = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:")
print(conf_matrix)   #### 해석?
```

```python

# 혼동 행렬 시각화
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", cbar=False)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```
