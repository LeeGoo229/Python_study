# 15. 엘보우 기법으로 최적 군집 수 찾기 (Elbow Method)

`make_blobs`로 4개 중심을 가진 가상 데이터를 생성하고, 군집 수(`k=2~10`)별 이너셔(`inertia_`) 변화를 시각화해 엘보우 지점을 탐색합니다.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
```

## 문제 1. k=2~10 범위의 이너셔(`inertia_`) 계산

`n_clusters`를 2부터 10까지 변화시키며 `KMeans(..., init='k-means++', max_iter=300, n_init=10, random_state=0)`를 학습하고 각 `k`별 `inertia_`를 출력하세요.

```python
inertia = []
for k in range(2, 11):
    km = KMeans(n_clusters=k, init='k-means++', max_iter=300, n_init=10, random_state=0)
    km.fit(X)
    inertia.append(km.inertia_)
    print(f'k={k}: inertia={km.inertia_:.2f}')
```

**결과 해석**

군집 수 `k`가 증가할수록 각 점과 중심점 간 거리 제곱합인 `inertia_`는 단조 감소하며, `k=4`까지는 급격히 줄다가 이후부터 감소 폭이 완만해집니다.

## 문제 2. 엘보우 그래프(Elbow Plot) 시각화

계산한 `inertia` 리스트를 `x=range(2, 11)`에 대해 선 그래프로 시각화하고 꺾이는 지점(팔꿈치)을 확인하세요.

```python
inertia = []
for k in range(2, 11):
    km = KMeans(n_clusters=k, init='k-means++', max_iter=300, n_init=10, random_state=0)
    km.fit(X)
    inertia.append(km.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(range(2, 11), inertia, marker='o')
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

**결과 해석**

그래프가 `k=4`에서 뚜렷하게 꺾이는 팔꿈치(Elbow) 형태를 보이므로, 이 데이터의 최적 군집 수가 4개임을 시각적으로 확인할 수 있습니다.
