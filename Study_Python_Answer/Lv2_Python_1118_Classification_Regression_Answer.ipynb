# 1. 분류와 회귀

분류는 범주를 예측하고 회귀는 수치를 예측합니다. Iris 분류와 주택가격 회귀의 지표를 비교합니다. 원본 California Housing 예제는 최초 실행 시 데이터 다운로드가 필요합니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## 문제 1. Iris 분류

로지스틱 회귀로 꽃 종류를 예측하고 테스트 정확도를 구하세요.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
model = LogisticRegression(max_iter=500).fit(X_train, y_train)
print(accuracy_score(y_test, model.predict(X_test)))
```

**결과 해석**

정확도는 예측한 범주가 실제 범주와 일치한 관측의 비율입니다.

## 문제 2. 당뇨 진행도 회귀

내장 Diabetes 데이터를 분할하고 선형 회귀의 테스트 MAE와 MSE를 구하세요.

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
diabetes = load_diabetes()
X_train, X_test, y_train, y_test = train_test_split(
    diabetes.data, diabetes.target, test_size=0.3, random_state=42
)
pred = LinearRegression().fit(X_train, y_train).predict(X_test)
print(mean_absolute_error(y_test, pred), mean_squared_error(y_test, pred))
```

**결과 해석**

MAE는 목표 변수와 같은 단위입니다. MSE는 오차를 제곱하므로 큰 오차에 더 민감합니다.
