# 5. 와인 데이터 탐색과 스케일링

Wine 데이터의 특성 분포와 클래스 비율을 탐색하고, 분할 후 표준화가 KNN에 적용되는 방식을 살펴봅니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
wine = load_wine(as_frame=True)
X, y = wine.data, wine.target
```

## 문제 1. 와인 데이터 탐색

알코올 특성의 분포와 클래스별 관측 비율을 확인하세요.

```python
import matplotlib.pyplot as plt
print(y.value_counts(normalize=True).sort_index())
X['alcohol'].plot.hist(bins=12, title='Wine alcohol distribution')
plt.show()
```

**결과 해석**

클래스 비율로 분할 시 층화가 필요한지 판단하고, 히스토그램으로 특성의 범위와 분포를 확인할 수 있습니다.

## 문제 2. 스케일링과 KNN

데이터를 층화 분할하고 StandardScaler와 KNN을 파이프라인으로 연결해 평가하세요.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

**결과 해석**

파이프라인은 훈련 데이터에서 스케일링 기준을 학습하고 테스트 데이터에 같은 기준을 적용합니다.
