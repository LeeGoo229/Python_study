# 8. 신경망 주택 가격 회귀

표준화와 MLPRegressor로 주택 가격을 예측합니다. 은닉층 크기와 실제 가중치 행렬을 살펴보고 소규모 데이터의 평가 한계를 확인합니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
from pathlib import Path
import pandas as pd

data_dir = DATA_DIR if 'DATA_DIR' in globals() else Path('data')
if not (data_dir / 'house.price.csv').is_file():
    data_dir = Path('Study_Python') / '[7] Machine_Learning' / 'data'
if not (data_dir / 'house.price.csv').is_file():
    data_dir = Path('..') / 'Study_Python' / '[7] Machine_Learning' / 'data'
file_path = data_dir / 'house.price.csv'
df = pd.read_csv(file_path)
from sklearn.model_selection import train_test_split

X = df[['year', 'size']]
y = df['price']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

## 문제 1. 신경망 회귀 파이프라인

훈련 기준 표준화와 은닉 노드 5개의 MLP를 파이프라인으로 연결하고 테스트 RMSE와 MAE를 구하세요.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
model = make_pipeline(StandardScaler(), MLPRegressor(
    hidden_layer_sizes=(5,), solver='lbfgs', max_iter=1000, random_state=42
))
model.fit(X_train, y_train)
pred = model.predict(X_test)
print(mean_squared_error(y_test, pred) ** 0.5, mean_absolute_error(y_test, pred))
print([weights.shape for weights in model.named_steps['mlpregressor'].coefs_])
```

**결과 해석**

입력 2개, 은닉 5개, 출력 1개이면 가중치 행렬 모양은 (2, 5), (5, 1)입니다.

## 문제 2. 평균 예측 기준선

훈련 목표값 평균으로 테스트 관측 전체를 예측하는 기준선의 RMSE와 MAE를 구하세요.

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
baseline = DummyRegressor(strategy='mean').fit(X_train, y_train)
pred = baseline.predict(X_test)
print(mean_squared_error(y_test, pred) ** 0.5, mean_absolute_error(y_test, pred))
```

**결과 해석**

복잡한 모델의 오차를 기준선과 비교하세요. 테스트 표본이 두 개라 결과의 변동성이 큽니다.
