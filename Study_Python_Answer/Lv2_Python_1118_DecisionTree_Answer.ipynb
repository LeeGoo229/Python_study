# 9. 의사결정나무 분류 기초 (Decision Tree)

Iris 데이터셋으로 의사결정나무 분류기를 학습하고 `max_depth`와 분할 기준(`gini` vs `entropy`)에 따른 성능 차이를 확인합니다.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)
```

## 문제 1. DecisionTreeClassifier 학습 및 정확도 확인

`DecisionTreeClassifier(random_state=42)` 모델을 훈련 데이터로 학습하고 테스트 데이터 정확도(`score`)와 트리 깊이(`get_depth()`)를 출력하세요.

```python
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)
acc = clf.score(X_test, y_test)
print(f'Accuracy: {acc*100:.2f}%, Depth: {clf.get_depth()}')
```

**결과 해석**

`max_depth=None` 기본 설정에서는 순수 노드가 될 때까지 트리가 깊어지며 테스트 데이터에서 높은 정확도를 보이지만, 깊이 제한이 없으면 과적합 위험이 커질 수 있습니다.

## 문제 2. max_depth 제한 및 분기 규칙 출력

`max_depth=2`, `criterion='entropy'`, `random_state=42`로 제한된 트리를 학습하고, 테스트 정확도와 `export_text` 분기 규칙을 출력하세요.

```python
clf_d2 = DecisionTreeClassifier(criterion='entropy', max_depth=2, random_state=42)
clf_d2.fit(X_train, y_train)
print(f'Accuracy (max_depth=2): {clf_d2.score(X_test, y_test)*100:.2f}%')
print(export_text(clf_d2, feature_names=iris.feature_names))
```

**결과 해석**

트리의 최대 깊이를 2로 제한해도 핵심 특성(`petal length`, `petal width`) 기준의 간결한 규칙만으로 Iris 품종을 효과적으로 분류할 수 있습니다.
