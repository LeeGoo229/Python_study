## 문제 1. 교육 과정 수료 여부 판정기 [조건문 실습]

학생의 `attendance`(출석률, %)와 `assignments`(과제 제출 횟수)를 기반으로 수료 여부를 판별하는 코드를 작성하세요.
- **수료 기준:** 출석률 80% 이상 **이고(AND)** 과제 제출 횟수 3회 이상인 경우 `"수료 가능"`
- 그 외의 경우 `"수료 불가능"`

테스트 케이스:
1. `att = 90, asm = 5` -> 결과: `수료 가능`
2. `att = 90, asm = 1` -> 결과: `수료 불가능`
3. `att = 75, asm = 5` -> 결과: `수료 불가능`

세 가지 경우를 각각 변수에 넣고 판정 결과를 출력하세요.


```python
# 문제 1 풀이
test_cases = [
    {"att": 90, "asm": 5},
    {"att": 90, "asm": 1},
    {"att": 75, "asm": 5}
]

for idx, tc in enumerate(test_cases, 1):
    att = tc["att"]
    asm = tc["asm"]
    if att >= 80 and asm >= 3:
        result = "수료 가능"
    else:
        result = "수료 불가능"
    print(f"케이스 {idx}: 출석률 {att}%, 과제 {asm}회 -> {result}")
```

**결과 해석 / 확인한 점:**  
1. **논리곱(AND) 연산의 엄격성:** `and` 연산자는 양쪽 조건이 모두 참(`True`)이어야 참을 반환합니다. 출석률이 90%로 충분해도 과제가 1회이면 기준에 미달하여 수료 불가능으로 정확히 판정됩니다.
2. **실무 적용:** 사용자 작성 실습 노트북(`[실습]_Python_if제어문_이희구작성_1103.ipynb`)의 핵심 요건을 충실히 반영하여 비즈니스 규칙 판정 로직을 자동화했습니다.


## 문제 2. 다중 고객 세그먼트 등급 분류 (if-elif-else) [비즈니스 로직]

누적 구매 금액(`spend`)에 따라 고객 등급을 4단계로 분류하려고 합니다.
- 1,000,000원 이상: `"VIP"`
- 500,000원 이상 ~ 1,000,000원 미만: `"GOLD"`
- 200,000원 이상 ~ 500,000원 미만: `"SILVER"`
- 200,000원 미만: `"BRONZE"`

setup의 `customer_data` 리스트를 for 문으로 순회하면서, 각 고객의 `user_id`와 부여된 `grade`를 출력하세요.


```python
# 문제 2 풀이
for c in customer_data:
    spend = c["spend"]
    if spend >= 1000000:
        grade = "VIP"
    elif spend >= 500000:
        grade = "GOLD"
    elif spend >= 200000:
        grade = "SILVER"
    else:
        grade = "BRONZE"
    print(f"고객 ID: {c['user_id']} | 누적 금액: {spend:,}원 | 등급: {grade}")
```

**결과 해석 / 확인한 점:**  
1. **다중 조건의 순차적 평가:** `if-elif-else` 구문은 위에서부터 순서대로 평가되므로, 가장 높은 기준값(100만 원 이상)부터 내림차순으로 배치해야 조건이 겹치지 않고 정확히 분류됩니다.
2. **고객 분석(CRM) 관점:** VIP 2명(ID 1, 5), SILVER 1명(ID 2), BRONZE 2명(ID 3, 4)으로 세분화되어 차별화된 타겟 마케팅을 적용할 수 있습니다.


## 문제 3. enumerate를 활용한 이탈 위험 고객 인덱스 추적 [탐색 제어]

고객의 출석률 목록이 다음과 같습니다:
`attendance_scores = [88, 92, 70, 65, 95, 58, 84]`

출석률이 **75% 미만**인 고객을 '이탈 위험 고객'으로 간주합니다.
`enumerate()`를 사용하여 출석률 75% 미만인 고객의 **위치 인덱스(0부터 시작)**와 **출석 점수**를 찾아 출력하세요.


```python
# 문제 3 풀이
attendance_scores = [88, 92, 70, 65, 95, 58, 84]

print("=== 이탈 위험 고객 목록 (기준: 75% 미만) ===")
for idx, score in enumerate(attendance_scores):
    if score < 75:
        print(f"고객 번호(인덱스): {idx}번 | 출석 점수: {score}%")
```

**결과 해석 / 확인한 점:**  
1. **인덱스 추적의 장점:** `enumerate()`는 별도의 카운터 변수(`i = 0; i += 1`) 없이도 원본 데이터에서의 위치 정보를 O(1)로 제공합니다.
2. **이상 탐지 결과:** 2번(70%), 3번(65%), 5번(58%) 등 총 3건의 저출석 이상 관측치가 성공적으로 탐지되었습니다.


## 문제 4. zip을 활용한 복수 리스트 동시 집계 및 목표 초과 달성률 [데이터 매핑]

지점별 이번 달 실제 매출과 목표 매출 리스트가 있습니다.
- 지점명: `branches = ["강남점", "홍대점", "판교점", "여의도점", "수원점"]`
- 실제 매출: `actual_sales = [120, 85, 140, 95, 110]` (단위: 백만 원)
- 목표 매출: `target_sales = [100, 90, 120, 100, 100]` (단위: 백만 원)

`zip()`을 사용하여 다음을 구하세요:
1. 각 지점의 달성률 (`actual_sales / target_sales * 100`)을 계산하고 지점명과 함께 출력하세요.
2. 목표를 100% 이상 달성한 지점의 수를 카운트하여 출력하세요.


```python
# 문제 4 풀이
branches = ["강남점", "홍대점", "판교점", "여의도점", "수원점"]
actual_sales = [120, 85, 140, 95, 110]
target_sales = [100, 90, 120, 100, 100]

achieved_count = 0
for branch, actual, target in zip(branches, actual_sales, target_sales):
    rate = (actual / target) * 100
    is_success = rate >= 100.0
    if is_success:
        achieved_count += 1
    status = "달성" if is_success else "미달성"
    print(f"{branch}: 목표 {target}M, 실적 {actual}M -> 달성률 {rate:.1f}% ({status})")

print(f"\n총 목표 달성 지점 수: {achieved_count}개 지점 (총 5개 중)")
```

**결과 해석 / 확인한 점:**  
1. **다중 컬렉션 병렬 순회:** `zip()`을 통해 서로 다른 세 개의 리스트(지점, 실적, 목표)를 행 단위(row-wise)로 손쉽게 묶어 연산할 수 있습니다.
2. **성과 분석:** 강남(120%), 판교(116.7%), 수원(110%) 3개 지점이 초과 달성하였으며, 홍대(94.4%)와 여의도(95.0%)는 소폭 미달성한 현황이 명확히 도출되었습니다.


## 문제 5. 리스트 컴프리헨션(List Comprehension)을 통한 이상값 필터링 [효율적 전처리]

센서 데이터 스트림에서 비정상적인 음수값이나 결측을 뜻하는 극단치(9999)가 혼입되었습니다:
`sensor_readings = [23.5, 24.1, -999.0, 22.8, 9999.0, 25.0, 0.0, 23.9]`

리스트 컴프리헨션을 사용하여:
1. `0` 초과이면서 `100` 미만인 유효 센서 측정치만 추출하여 `valid_readings` 리스트를 생성하세요.
2. 정제된 리스트의 평균값(소수점 둘째 자리)을 구하여 출력하세요.


```python
# 문제 5 풀이
sensor_readings = [23.5, 24.1, -999.0, 22.8, 9999.0, 25.0, 0.0, 23.9]

# 1. 컴프리헨션을 이용한 필터링 (0 < x < 100)
valid_readings = [x for x in sensor_readings if 0 < x < 100]
print("유효 센서 측정값:", valid_readings)

# 2. 유효 데이터 평균 계산
avg_val = sum(valid_readings) / len(valid_readings)
print(f"정제된 센서 평균 온도: {avg_val:.2f}°C")
```

**결과 해석 / 확인한 점:**  
1. **컴프리헨션의 간결성과 속도:** 4줄 이상의 `for - if - append` 블록을 단 한 줄 `[x for x in data if cond]`로 축약하여 코드 가독성과 실행 속도를 모두 높였습니다.
2. **데이터 클렌징 효과:** 극단치(`-999.0`, `9999.0`)와 영점 오류(`0.0`)를 제거한 결과 실측 온도 평균이 약 23.86°C로 안정적인 수렴치를 보입니다.


## 문제 6. try-except 예외 처리를 통한 비정형 가격 데이터 수치화 [견고한 파이프라인]

setup 코드의 `raw_prices`에는 정수 문자열 외에도 `"null"`, `"SOLD_OUT"` 등 수치 변환 시 오류를 유발하는 문자열이 포함되어 있습니다:
- `raw_prices = ["12000", "45000", "null", "35000", "SOLD_OUT", "89000", "0"]`

요구사항:
1. for 루프와 `try-except (ValueError)`를 사용하여, 정수로 변환 가능한 값만 `clean_prices` 리스트에 정수형(`int`)으로 추가하세요.
2. 오류가 발생한 항목은 에러 메시지를 출력하며 건너뛰도록 하세요.
3. 변환된 순수 상품 가격들의 총합과 평균 가격을 출력하세요.


```python
# 문제 6 풀이
clean_prices = []
failed_items = []

for item in raw_prices:
    try:
        price = int(item)
        clean_prices.append(price)
    except ValueError:
        failed_items.append(item)
        print(f"[예외 감지] 수치 변환 불가 데이터 건너뜀: '{item}'")

print("\n변환 성공 가격 리스트:", clean_prices)
total_sum = sum(clean_prices)
avg_price = total_sum / len(clean_prices)
print(f"유효 가격 총액: {total_sum:,}원 | 평균 가격: {avg_price:,.1f}원")
```

**결과 해석 / 확인한 점:**  
1. **배치 작업 중단 방지:** 예외 처리가 없을 경우 세 번째 원소 `"null"`에서 `ValueError`가 발생하여 전체 프로그램이 중단되지만, `try-except`로 오류를 격리하여 배치 작업을 끝까지 완수할 수 있습니다.
2. **데이터 파이프라인 감사(Audit):** 변환 실패한 데이터(`failed_items`)를 따로 기록해둠으로써 원천 데이터 품질 개선을 위한 후속 조치가 가능합니다.
