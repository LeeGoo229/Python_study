# 11. 랜덤 포레스트 분류 (Random Forest)

Iris 데이터셋에 `RandomForestClassifier`를 적용하고 `max_features` 계산과 특성 중요도를 확인합니다.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)
```

## 문제 1. RandomForestClassifier 학습 및 max_features 계산

`RandomForestClassifier(n_estimators=100, random_state=42)`를 학습해 테스트 정확도와 기본 `max_features='sqrt'`에 해당하는 특성 수(`int(np.sqrt(X_train.shape[1]))`)를 출력하세요.

```python
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train)
pred = rf_clf.predict(X_test)
acc = accuracy_score(y_test, pred)
max_feat_cnt = int(np.sqrt(X_train.shape[1]))
print(f'Accuracy: {acc*100:.2f}%, sqrt(n_features): {max_feat_cnt}')
```

**결과 해석**

분류용 랜덤 포레스트는 각 노드 분할 시 전체 특성 중 제곱근 개수(`sqrt(4)=2`)만 무작위 후보로 고려해 트리 간 다양성을 확보하고 과적합을 줄입니다.

## 문제 2. 특성 중요도 비교

학습된 랜덤 포레스트의 `feature_importances_`를 DataFrame 또는 Series로 내림차순 정렬하여 가장 중요한 특성을 확인하세요.

```python
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train)
imp = pd.Series(rf_clf.feature_importances_, index=iris.feature_names).sort_values(ascending=False)
print(imp.round(4))
```

**결과 해석**

단일 트리보다 여러 트리의 중요도를 평균 내므로 특성 중요도 추정이 안정적이며, 꽃잎 길이/너비 특성이 분류에 가장 크게 기여함을 확인할 수 있습니다.
