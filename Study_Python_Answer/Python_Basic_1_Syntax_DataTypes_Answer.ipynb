## 문제 2. 부동소수점 연산과 f-string 서식 지정 [필수 기초]

한 이커머스에서 할인율을 적용하여 최종 결제액을 계산하려고 합니다.
- 정가: `original_price = 158000`
- 할인율: `discount_rate = 0.15` (15%)
- 부가세: 결제액의 10% (`vat_rate = 0.10`)

할인된 금액에 부가세를 더한 최종 결제 금액(`final_price`)을 계산하고,
f-string을 사용하여 소수점 둘째 자리까지 반올림된 형태로 다음과 같이 출력하세요.
출력 형식: `최종 결제 금액: 147,730.00원 (할인액: 23,700원)`


```python
# 문제 2 풀이
original_price = 158000
discount_rate = 0.15
vat_rate = 0.10

discount_amount = original_price * discount_rate
price_after_discount = original_price - discount_amount
final_price = price_after_discount * (1 + vat_rate)

print(f"최종 결제 금액: {final_price:,.2f}원 (할인액: {discount_amount:,.0f}원)")
```

**결과 해석 / 확인한 점:**  
1. **f-string 포맷 지정자 활용:** `{final_price:,.2f}` 문법을 통해 천 단위 콤마(`,`)와 고정 소수점 둘째 자리(`.2f`)를 손쉽게 표현할 수 있습니다.
2. **비즈니스 계산 검증:** 158,000원에서 15% 할인(23,700원)된 134,300원에 부가세 10%(13,430원)가 가산되어 최종 147,730.00원이 도출됨을 검증했습니다.


## 문제 3. 리스트 인덱싱, 슬라이싱과 메서드 조작 [필수 기초]

`weekly_visitors = [1200, 1350, 980, 1500, 1800, 2400, 2100]`은 월요일부터 일요일까지의 일일 방문자 수입니다.
다음 단계를 순서대로 수행하세요:
1. 주중(월 ~ 금, 인덱스 0 ~ 4) 방문자 수만 슬라이싱하여 `weekday_visitors`에 저장하고 출력하세요.
2. 주말(토, 일) 방문자 수의 평균을 계산하여 출력하세요.
3. 다음 주 월요일 예상 방문자 수 `1400`을 리스트의 가장 뒤에 `append()`하고 전체 리스트 길이를 확인하세요.


```python
# 문제 3 풀이
weekly_visitors = [1200, 1350, 980, 1500, 1800, 2400, 2100]

# 1. 주중 방문자 슬라이싱 (월~금: 인덱스 0, 1, 2, 3, 4)
weekday_visitors = weekly_visitors[:5]
print("주중 방문자:", weekday_visitors)

# 2. 주말 방문자 평균 (토, 일: 인덱스 5, 6)
weekend_visitors = weekly_visitors[5:]
weekend_avg = sum(weekend_visitors) / len(weekend_visitors)
print(f"주말 평균 방문자: {weekend_avg:.1f}명")

# 3. 신규 값 추가 및 길이 확인
weekly_visitors.append(1400)
print("업데이트된 전체 방문자 리스트:", weekly_visitors)
print("총 일수:", len(weekly_visitors))
```

**결과 해석 / 확인한 점:**  
1. **슬라이싱의 끝 인덱스 규칙:** `[:5]`는 0부터 4까지 총 5개의 원소를 포함합니다.
2. **시계열 패턴 관찰:** 주중 방문자 평균(약 1,366명)에 비해 주말 방문자 평균(2,250명)이 현저히 높은 주말 집중형 트래픽 패턴을 직관적으로 확인할 수 있습니다.


## 문제 4. 딕셔너리를 활용한 상품 메타데이터 관리 [데이터 조작]

딕셔너리는 DB 테이블의 단일 레코드나 JSON API 응답을 표현할 때 가장 널리 쓰입니다.
다음 정보를 담은 `product_info` 딕셔너리를 작성하고 요구사항을 구현하세요.
- `sku`: `"SKU-8821"`
- `name`: `"Wireless Mechanical Keyboard"`
- `price`: `89000`
- `stock`: `15`

요구사항:
1. `stock` 수량을 5개 감소(판매 반영)시키세요.
2. 새 항목 `"category": "Peripherals"`를 추가하세요.
3. 존재하지 않는 키 `"discount_coupon"`을 조회할 때 에러가 발생하지 않도록 `.get()` 메서드를 쓰고, 기본값 `"NONE"`이 반환되도록 하세요.


```python
# 문제 4 풀이
product_info = {
    "sku": "SKU-8821",
    "name": "Wireless Mechanical Keyboard",
    "price": 89000,
    "stock": 15
}

# 1. 재고 5개 감소
product_info["stock"] -= 5

# 2. 신규 카테고리 추가
product_info["category"] = "Peripherals"

# 3. .get() 메서드로 안전한 키 조회
coupon = product_info.get("discount_coupon", "NONE")

print("수정된 상품 정보:", product_info)
print("쿠폰 조회 결과:", coupon)
```

**결과 해석 / 확인한 점:**  
1. **딕셔너리의 가변성(Mutablity):** 기존 키에 새로운 값을 대입하면 덮어씌워지고, 새로운 키를 할당하면 즉시 구조가 확장됩니다.
2. **안전한 조회 기법:** 대괄호 접근(`dict['key']`)은 키가 없을 경우 `KeyError`로 프로그램이 중단되지만, `.get(key, default)`를 활용하면 데이터 결측 상황에서도 안전하게 기본값을 핸들링할 수 있습니다.


## 문제 5. 세트(Set)를 활용한 유입 고객 중복 제거 및 이탈 분석 [데이터 분석 응용]

setup 코드에 정의된 `customer_names` 리스트에서 중복 방문자를 집계하려고 합니다.
- `customer_names = ["Kim", "Lee", "Park", "Choi", "Lee", "Kang", "Kim"]`

또한, 지난 달 구매 고객 세트가 다음과 같을 때:
- `last_month_buyers = {"Kim", "Lee", "Park", "Yoon", "Song"}`

다음 결과를 구하세요:
1. 이번 달 방문한 순수 고유 고객(Unique) 목록과 고유 고객 수를 구하세요.
2. 지난 달 구매자 중 이번 달에도 방문한 '유지 고객(교집합)'을 구하세요.
3. 지난 달 구매자였으나 이번 달에 방문하지 않은 '이탈 고객(차집합)'을 구하세요.


```python
# 문제 5 풀이
this_month_visitors_set = set(customer_names)
last_month_buyers = {"Kim", "Lee", "Park", "Yoon", "Song"}

# 1. 고유 고객 수
unique_count = len(this_month_visitors_set)
print("이번 달 고유 방문 고객:", this_month_visitors_set)
print("고유 고객 수:", unique_count)

# 2. 유지 고객 (교집합: &)
retained_customers = last_month_buyers & this_month_visitors_set
print("유지 고객 (교집합):", retained_customers)

# 3. 이탈 고객 (차집합: -)
churned_customers = last_month_buyers - this_month_visitors_set
print("이탈 고객 (차집합):", churned_customers)
```

**결과 해석 / 확인한 점:**  
1. **중복 제거(Deduplication) 효율성:** `set()`은 해시 테이블 기반으로 구현되어 있어 O(N)의 시간으로 수백만 건의 로그 데이터에서도 즉시 유니크 고객을 추출합니다.
2. **리텐션 분석의 기초:** 교집합 연산(`&`)은 재방문/유지 유저를 식별하고, 차집합 연산(`-`)은 프로모션 대상이 될 이탈(Churn) 고객 타겟팅에 직결됩니다.


## 문제 6. 문자열 정제와 형 변환(Type Casting) 파이프라인 [실무 전처리]

setup 코드의 `sample_order_raw` 변수는 웹 로그에서 수집된 비정형 주문 문자열입니다:
- `sample_order_raw = "  ORD_20261116_KR_#9901  "`

다음 단계로 문자열을 정제하세요:
1. 앞뒤 공백을 제거(`strip()`)하세요.
2. 언더바(`_`)를 기준으로 문자열을 분리(`split()`)하여 리스트로 만드세요.
3. 4번째 요소인 주문번호(`"#9901"`)에서 특수문자 `#`를 제거하고 순수 정수(`int`)로 변환하여 출력하세요.


```python
# 문제 6 풀이
# 1. 앞뒤 공백 제거
cleaned_str = sample_order_raw.strip()

# 2. 구분자 '_' 기준 분리
tokens = cleaned_str.split("_")
print("분리된 토큰 리스트:", tokens)

# 3. 4번째 요소(#9901)에서 '#' 제거 후 정수 변환
raw_id = tokens[3]
clean_id_int = int(raw_id.replace("#", ""))

print(f"최종 추출된 주문 번호: {clean_id_int} (타입: {type(clean_id_int)})")
```

**결과 해석 / 확인한 점:**  
1. **실무 텍스트 파싱 파이프라인:** 웹 로그나 CSV 파일에서 흔히 발생하는 불필요한 공백과 구분자, 특수문자를 정제하고 연산 가능한 숫자형(`int`)으로 변환하는 전형적인 ETL 과정을 완수했습니다.
2. **형 변환 안정성:** `#`와 같은 문자가 남아있는 상태에서 `int()`를 호출하면 `ValueError`가 발생하므로 전처리 순서가 매우 중요합니다.
