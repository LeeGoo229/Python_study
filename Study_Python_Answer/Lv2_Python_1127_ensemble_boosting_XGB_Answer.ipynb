# 27. XGBoost 부스팅 분류 (XGBoost)

L1/L2 규제와 병렬 트리 구축을 지원하는 `xgboost.XGBClassifier`를 학습하고 평가 지표(`eval_metric`) 설정과 분류 정확도를 확인합니다.

```python
import xgboost as xgb
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.3, random_state=42, stratify=iris.target
)
```

## 문제 1. XGBClassifier 학습 및 테스트 정확도 확인

`xgb.XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, eval_metric="mlogloss", random_state=42)`를 학습하고 테스트 정확도를 출력하세요.

```python
xgb_clf = xgb.XGBClassifier(
    n_estimators=100, learning_rate=0.1, max_depth=3, eval_metric='mlogloss', random_state=42
)
xgb_clf.fit(X_train, y_train)
print(f'XGBoost Accuracy: {accuracy_score(y_test, xgb_clf.predict(X_test)):.4f}')
```

**결과 해석**

XGBoost는 정규화(L1/L2)와 레벨 중심 트리 분할(Level-wise)을 통해 안정적인 다중 클래스 분류 성능을 제공합니다.

## 문제 2. XGBoost 특성 중요도 확인하기

학습된 `xgb_clf`의 `feature_importances_`를 Iris 특성 이름과 함께 출력하세요.

```python
xgb_clf = xgb.XGBClassifier(
    n_estimators=100, learning_rate=0.1, max_depth=3, eval_metric='mlogloss', random_state=42
)
xgb_clf.fit(X_train, y_train)

for name, imp in zip(iris.feature_names, xgb_clf.feature_importances_):
    print(f'{name}: {imp:.4f}')
```

**결과 해석**

XGBoost에서도 꽃잎(petal) 관련 특성이 꽃받침(sepal) 특성보다 높은 분기 기여도를 나타냅니다.
