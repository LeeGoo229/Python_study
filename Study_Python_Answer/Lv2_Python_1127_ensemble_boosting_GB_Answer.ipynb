# 25. 그래디언트 부스팅 분류 (Gradient Boosting)

이전 트리의 잔여 오차(Gradient)를 순차적으로 보정하는 `GradientBoostingClassifier`를 학습하고 `learning_rate`와 `max_depth` 변화에 따른 정확도를 비교합니다.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.3, random_state=42, stratify=iris.target
)
```

## 문제 1. GradientBoostingClassifier 학습 및 테스트 정확도 확인

`GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)`를 학습하고 테스트 정확도를 출력하세요.

```python
gb_clf = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
gb_clf.fit(X_train, y_train)
print(f'GB Accuracy: {accuracy_score(y_test, gb_clf.predict(X_test)):.4f}')
```

**결과 해석**

그래디언트 부스팅은 손실 함수의 기울기 방향으로 약한 트리를 누적 학습해 정형 데이터에서 높은 예측력을 발휘합니다.

## 문제 2. 특성 중요도(feature_importances_) 출력하기

학습된 `gb_clf`의 `feature_importances_`를 Iris 특성 이름(`iris.feature_names`)과 함께 출력하세요.

```python
gb_clf = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
gb_clf.fit(X_train, y_train)

for name, imp in zip(iris.feature_names, gb_clf.feature_importances_):
    print(f'{name}: {imp:.4f}')
```

**결과 해석**

부스팅 트리의 특성 중요도를 통해 꽃잎 길이와 너비가 품종 분류에 핵심적인 기여를 함을 확인할 수 있습니다.
