# 22. 에이다부스트 앙상블 분류 (AdaBoost)

오분류된 샘플의 가중치를 단계적으로 높여 약한 학습기를 결합하는 `AdaBoostClassifier`를 학습하고 기반 트리 깊이(`max_depth`)에 따른 성능 변화를 비교합니다.

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X, y = make_classification(n_samples=1000, n_features=20, n_informative=2, n_redundant=0, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## 문제 1. 기본 AdaBoostClassifier 학습 및 정확도 확인

`AdaBoostClassifier(n_estimators=100, random_state=42)`를 학습해 테스트 정확도를 출력하세요.

```python
ada_clf = AdaBoostClassifier(n_estimators=100, random_state=42)
ada_clf.fit(X_train, y_train)
print(f'AdaBoost (default) Accuracy: {accuracy_score(y_test, ada_clf.predict(X_test)):.4f}')
```

**결과 해석**

깊이 1의 결정 스텀프를 순차적으로 부스팅하는 기본 AdaBoost만으로도 오분류 샘플 가중치 보정을 통해 높은 분류 정확도를 달성합니다.

## 문제 2. 기반 트리 깊이와 학습률 조정 비교

`estimator=DecisionTreeClassifier(max_depth=3, random_state=42)`, `n_estimators=100`, `learning_rate=0.5`로 AdaBoost를 학습하고 정확도를 비교 출력하세요.

```python
ada_deep = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=3, random_state=42),
    n_estimators=100,
    learning_rate=0.5,
    random_state=42,
)
ada_deep.fit(X_train, y_train)
print(f'AdaBoost (depth=3, lr=0.5) Accuracy: {accuracy_score(y_test, ada_deep.predict(X_test)):.4f}')
```

**결과 해석**

기반 학습기가 복잡해질수록 표현력은 높아지지만 과적합 위험이 생기므로 `learning_rate`를 낮춰 균형을 맞추는 것이 중요합니다.
