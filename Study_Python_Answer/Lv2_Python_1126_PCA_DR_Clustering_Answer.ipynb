# 21. PCA 차원 축소와 군집·분류 비교 (PCA Dimensionality Reduction)

고차원(50차원) 데이터에서 PCA 2차원 축소 전후의 KMeans 실루엣 스코어를 비교하고, Iris 데이터에서 데이터 누수 없는 `Pipeline(StandardScaler + PCA + KNN)` 분류 성능을 확인합니다.

```python
from sklearn.datasets import make_classification, load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier

X_hd, _ = make_classification(
    n_samples=300, n_features=50, n_informative=5, n_redundant=0, n_clusters_per_class=1, random_state=42
)
X_scaled = StandardScaler().fit_transform(X_hd)
```

## 문제 1. 50차원 vs PCA 2차원 KMeans 실루엣 스코어 비교하기

`X_scaled`(50차원)와 `PCA(n_components=2, random_state=42)`로 축소한 `X_pca`에 각각 `KMeans(n_clusters=3, n_init=10, random_state=42)`를 적용해 두 실루엣 스코어를 비교 출력하세요.

```python
X_pca = PCA(n_components=2, random_state=42).fit_transform(X_scaled)
labels_hd = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(X_scaled)
labels_pca = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(X_pca)
print(f'50D Silhouette:    {silhouette_score(X_scaled, labels_hd):.4f}')
print(f'PCA-2D Silhouette: {silhouette_score(X_pca, labels_pca):.4f}')
```

**결과 해석**

고차원에 노이즈 특성이 많을수록 거리 기반 군집의 분리도가 떨어지며, PCA로 핵심 분산 방향만 남기면 군집 응집도와 분리도가 크게 개선됩니다.

## 문제 2. Iris 데이터 누수 방지 PCA + KNN 파이프라인 학습하기

Iris 데이터를 `test_size=0.2, stratify=y, random_state=42`로 먼저 분할한 뒤 `Pipeline([StandardScaler, PCA(n_components=2), KNeighborsClassifier(n_neighbors=5)])`의 테스트 정확도를 출력하세요.

```python
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42, stratify=iris.target
)
pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA(n_components=2, random_state=42)),
    ('knn', KNeighborsClassifier(n_neighbors=5)),
])
pipe.fit(X_train, y_train)
print(f'Iris PCA-2D + KNN Accuracy: {accuracy_score(y_test, pipe.predict(X_test)):.4f}')
```

**결과 해석**

데이터 분할 후 파이프라인 내부에서 스케일링과 PCA를 훈련 데이터에만 적합(fit)하면 테스트 정보 누수 없이 차원 축소 분류 성능을 평가할 수 있습니다.
