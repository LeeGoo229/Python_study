# 1. 혼동 행렬 기초와 시각화 (Confusion Matrix 1)

`confusion_matrix`를 계산하고 TN, FP, FN, TP 각 항목을 분해하여 해석합니다.

```python
from sklearn.metrics import confusion_matrix

y_true = [1, 0, 0, 1, 1, 0, 1, 0]
y_pred = [1, 1, 0, 1, 0, 0, 1, 0]
```

## 문제 1. 혼동 행렬 계산 및 TN, FP, FN, TP 추출

`confusion_matrix(y_true, y_pred)`를 계산하고 `ravel()`을 이용해 `tn, fp, fn, tp`를 각각 출력하세요.

```python
cm = confusion_matrix(y_true, y_pred)
tn, fp, fn, tp = cm.ravel()
print('Confusion Matrix:\n', cm)
print(f'TN={tn}, FP={fp}, FN={fn}, TP={tp}')
```

**결과 해석**

행은 실제 클래스(0, 1), 열은 예측 클래스(0, 1)를 의미하며 대각선 성분(TN=3, TP=3)이 올바르게 맞힌 샘플 수입니다.

## 문제 2. 정분류율(Accuracy)과 오분류 수 확인

추출한 `tn, fp, fn, tp`를 이용해 직접 정확도 `(tp + tn) / (tn + fp + fn + tp)`와 오분류 건수 `fp + fn`을 계산하세요.

```python
cm = confusion_matrix(y_true, y_pred)
tn, fp, fn, tp = cm.ravel()
acc = (tp + tn) / (tn + fp + fn + tp)
print(f'Accuracy: {acc:.2f}, Misclassified: {fp + fn}')
```

**결과 해석**

전체 8개 중 6개를 맞혀 정확도는 0.75이며, FP(1건)와 FN(1건) 총 2건의 오분류가 발생했습니다.
