# 14. K-평균 군집화와 Iris 품종 비교 (K-Means Clustering)

레이블 없이 Iris 데이터의 4개 특성만으로 `KMeans` 군집화를 수행하고, 할당된 `Cluster_ID`와 실제 품종(`target`)을 교차표로 비교합니다.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

iris = load_iris()
X = iris.data
```

## 문제 1. KMeans 군집화 실행 및 DataFrame 결합

`KMeans(n_clusters=3, init='k-means++', max_iter=300, n_init=10, random_state=0)`로 `X`를 군집화(`fit_predict`)하고, `iris_df`에 `Cluster_ID` 열을 추가해 상위 5행을 출력하세요.

```python
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, n_init=10, random_state=0)
y_kmeans = kmeans.fit_predict(X)
iris_df = pd.DataFrame(X, columns=iris.feature_names)
iris_df['Cluster_ID'] = y_kmeans
print(iris_df.head())
```

**결과 해석**

비지도 학습인 `KMeans`는 정답 레이블 없이도 특성 간 유클리드 거리를 기반으로 150개 샘플을 3개의 군집(`0`, `1`, `2`)으로 분할합니다.

## 문제 2. 실제 품종과 군집 결과 교차표(`pd.crosstab`) 비교

`iris_df['target'] = iris.target`을 추가한 뒤 `pd.crosstab(iris_df['target'], iris_df['Cluster_ID'])`를 출력해 군집 일치도를 확인하세요.

```python
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, n_init=10, random_state=0)
y_kmeans = kmeans.fit_predict(X)
iris_df = pd.DataFrame(X, columns=iris.feature_names)
iris_df['Cluster_ID'] = y_kmeans
iris_df['target'] = iris.target
ct = pd.crosstab(iris_df['target'], iris_df['Cluster_ID'])
print(ct)
```

**결과 해석**

Setosa(`target=0`)는 50개 모두 단일 군집으로 완벽히 분리되며, Versicolor(`1`)와 Virginica(`2`)는 특성 분포가 일부 겹쳐 소수 샘플이 인접 군집에 배정됩니다.
