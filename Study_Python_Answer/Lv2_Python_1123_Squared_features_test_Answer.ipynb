# 12. 특성 변환과 선형 분리 (Squared Features)

직선으로 나눌 수 없는 비선형 데이터를 제곱 특성 변환($x_1^2$)으로 선형 분리가 가능한 공간으로 매핑하는 원리를 시각화합니다.

```python
import matplotlib.pyplot as plt

x1 = [-1, 1, -3, 3]
x2 = [1, 1, 1, 1]
colors = ['blue', 'blue', 'red', 'red']
```

## 문제 1. 리스트 컴프리헨션으로 제곱 특성 변환하기

`x1`의 각 원소를 제곱한 `x1_squared` 리스트를 생성하고, 변환 전 `x1`과 변환 후 `x1_squared`를 출력하세요.

```python
x1_squared = [x ** 2 for x in x1]
print('변환 전 x1:', x1)
print('변환 후 x1_squared:', x1_squared)
```

**결과 해석**

`[-1, 1, -3, 3]`이 `[1, 1, 9, 9]`로 변환되면서 파란색 점(`1`)과 빨간색 점(`9`)이 1차원 축에서 명확히 분리됩니다.

## 문제 2. 변환 전/후 산점도와 분리 경계선 시각화

변환 전 `(x1, x2)`와 변환 후 `(x1_squared, x2)` 산점도를 좌우 서브플롯(`plt.subplot(1, 2, 1/2)`)으로 그리고, 변환 후 그래프에 `x=5` 기준 수직선(`plt.axvline`)을 추가하세요.

```python
x1_squared = [x ** 2 for x in x1]
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter(x1, x2, c=colors, s=100)
plt.title('Before')
plt.xlabel('x1')
plt.ylabel('x2')
plt.xlim(-4, 4)

plt.subplot(1, 2, 2)
plt.scatter(x1_squared, x2, c=colors, s=100)
plt.axvline(x=5, color='black', linestyle='--')
plt.title('After: x1 squared')
plt.xlabel('x1^2')
plt.xlim(0, 10)

plt.tight_layout()
plt.show()
```

**결과 해석**

원래 공간에서는 직선 하나로 두 그룹을 나눌 수 없지만, 비선형 특성 변환($x_1^2$) 이후에는 `x=5` 직선 하나로 완벽하게 분류할 수 있습니다.
