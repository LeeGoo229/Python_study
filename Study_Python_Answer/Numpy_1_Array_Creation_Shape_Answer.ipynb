## 문제 1. ndarray 생성과 단일 데이터 타입 업캐스팅 규칙 [자료형 검증]

정수와 부동소수점이 섞인 `sample_raw_list = [10, 20.5, 30, 40.8, 50]`이 있습니다.
1. `np.array()`를 사용하여 NumPy 배열 `arr_float`로 변환하세요.
2. `arr_float.dtype`을 확인하여 어떤 타입으로 자동 통일되었는지 출력하세요.
3. `.astype(np.int32)` 메서드를 사용하여 강제로 32비트 정수형으로 형변환한 `arr_int`를 만들고 소수점 절삭 결과를 출력하세요.


````python
# 문제 1 풀이
sample_raw_list = [10, 20.5, 30, 40.8, 50]

# 1. NumPy 배열 생성
arr_float = np.array(sample_raw_list)
print("생성된 배열:", arr_float)

# 2. 자동 업캐스팅 확인
print("자동 할당된 dtype:", arr_float.dtype)

# 3. int32로 명시적 형변환 (소수점 버림)
arr_int = arr_float.astype(np.int32)
print("정수 변환 배열:", arr_int)
print("변환 후 dtype:", arr_int.dtype)
````

**결과 해석 / 확인한 점:**  
1. **타입 통일 규칙(Type Promotion):** 파이썬 리스트는 요소마다 다른 타입을 담을 수 있지만, NumPy는 고속 C 배열이므로 정수(`10`)가 실수(`10.0`)로 승격되어 `float64`로 통일됩니다.
2. **형변환 주의점:** `astype(np.int32)` 적용 시 반올림이 아니라 소수점 이하가 버려져(Truncation) `20.5`가 `20`으로, `40.8`이 `40`으로 변환됨을 확인했습니다.


## 문제 2. 수치 배열 생성: np.arange vs np.linspace [배열 생성]

수치 시각화나 회귀선 X축을 그릴 때 두 함수가 자주 쓰입니다.
1. `np.arange()`를 사용하여 10부터 50까지 5 간격으로 증가하는 배열 `arr_step`을 생성하세요 (50 포함).
2. `np.linspace()`를 사용하여 0부터 100까지 정확히 11개의 균등 분할 점을 갖는 배열 `arr_points`를 생성하세요.
3. 두 배열의 길이(`len`)와 원소들을 각각 출력하세요.


````python
# 문제 2 풀이
# 1. np.arange: [시작, 끝, 간격) - 50을 포함하려면 stop에 51 지정
arr_step = np.arange(10, 51, 5)
print("arange 결과 (길이:", len(arr_step), "):", arr_step)

# 2. np.linspace: [시작, 끝, 총 개수] - 끝점 100 포함
arr_points = np.linspace(0, 100, 11)
print("linspace 결과 (길이:", len(arr_points), "):", arr_points)
````

**결과 해석 / 확인한 점:**  
1. **간격 지정 vs 개수 지정:** `np.arange`는 사용자가 '간격(Step=5)'을 알고 있을 때 쓰고, `np.linspace`는 시작과 끝 사이를 정확히 'N개(11개)' 구간으로 등분하고자 할 때 사용합니다.
2. **경계값 포함 여부:** `arange`는 stop(51) 미만이지만, `linspace`는 기본적으로 stop(100)을 포함하여 0, 10, 20, ..., 100이 생성됩니다.


## 문제 3. 다차원 배열의 메타데이터 속성 4종 세트 분석 [구조 진단]

setup 코드에 정의된 `matrix_4x5`는 1부터 20까지 채워진 2차원 행렬입니다.
다음을 출력하여 배열의 구조를 점검하세요:
1. 차원의 수 (`ndim`)
2. 행과 열의 크기 (`shape`)
3. 전체 원소 개수 (`size`)
4. 각 원소 하나의 바이트 크기 (`itemsize`) 및 전체 메모리 사용량 (`nbytes`)


````python
# 문제 3 풀이
print("대상 행렬:\n", matrix_4x5)
print("-" * 30)
print(f"1. 차원 (ndim): {matrix_4x5.ndim}차원")
print(f"2. 형상 (shape): {matrix_4x5.shape} (4행 5열)")
print(f"3. 전체 원소 수 (size): {matrix_4x5.size}개")
print(f"4. 단일 원소 바이트 (itemsize): {matrix_4x5.itemsize} bytes")
print(f"5. 총 메모리 사용량 (nbytes): {matrix_4x5.nbytes} bytes")
````

**결과 해석 / 확인한 점:**  
1. **메모리 계산식 검증:** `size(20)` * `itemsize(8 bytes for int64)` = `160 bytes`로, 배열 전체가 160바이트의 연속된 메모리를 점유함을 확인했습니다.
2. **차원 불일치 디버깅:** 딥러닝 텐서나 행렬 연산 전 `.shape` 확인은 차원 불일치(Shape Mismatch)를 방지하는 가장 중요한 습관입니다.


## 문제 4. 2차원 배열 인덱싱과 서브 매트릭스 슬라이싱 [슬라이싱 마스터]

`matrix_4x5`에서 다음 조건에 맞는 요소들을 슬라이싱하여 출력하세요:
1. 2행 3열에 위치한 단일 원소 (0부터 시작하므로 2행 3열)
2. 1번째 행 전체 (2번째 줄 전체)
3. 4번째 열(인덱스 3) 전체
4. 1~2행(2, 3번째 행) 및 1~3열(2, 3, 4번째 열)에 해당하는 2x3 크기의 중심 서브 매트릭스


````python
# 문제 4 풀이
# 1. 2행 3열 단일 원소
elem_2_3 = matrix_4x5[2, 3]
print("1. 2행 3열 원소:", elem_2_3)

# 2. 1번째 행 전체
row_1 = matrix_4x5[1, :]
print("2. 1번째 행 전체:", row_1)

# 3. 4번째 열(인덱스 3) 전체
col_3 = matrix_4x5[:, 3]
print("3. 인덱스 3열 전체:", col_3)

# 4. 2x3 서브 매트릭스 [1:3, 1:4]
sub_mat = matrix_4x5[1:3, 1:4]
print("4. 중심 2x3 서브 매트릭스:\n", sub_mat)
````

2. **서브 매트릭스 검증:** 1~2행과 1~3열의 교차 영역으로 `[[7, 8, 9], [12, 13, 14]]`가 정확히 추출되었습니다.


## 문제 5. reshape와 -1 와일드카드를 활용한 다차원 재배열 [형상 변환]

총 24개의 원소를 가진 1차원 센서 데이터 `sensor_1d = np.arange(101, 125)`가 있습니다.
1. `sensor_1d`를 4행 6열(`(4, 6)`)의 2차원 행렬로 변환하세요.
2. `-1` 와일드카드를 사용하여 열의 크기를 `8`로 고정하고 행은 자동 계산되도록 변환한 뒤 shape를 확인하세요.
3. 2행 3열 배열 4장으로 구성된 3차원 텐서(`(4, 2, 3)`)로 변환하고 첫 번째 장(인덱스 0)을 출력하세요.


````python
# 문제 5 풀이
sensor_1d = np.arange(101, 125)
print("원래 1차원 크기:", sensor_1d.shape)

# 1. 4행 6열 변환
mat_4x6 = sensor_1d.reshape(4, 6)
print("1. 4x6 형상:", mat_4x6.shape)

# 2. -1 와일드카드 활용 (열 8개 -> 24/8 = 3행 자동 계산)
mat_auto = sensor_1d.reshape(-1, 8)
print(f"2. 자동 계산된 형상: {mat_auto.shape} (3행 8열)")

# 3. 3차원 텐서 (4, 2, 3)
tensor_3d = sensor_1d.reshape(4, 2, 3)
print(f"3. 3차원 텐서 형상: {tensor_3d.shape}")
print("첫 번째 장 (장 인덱스 0):\n", tensor_3d[0])
````

**결과 해석 / 확인한 점:**  
1. **총 원소 수 불변의 법칙:** 24개의 원소는 `4x6`, `3x8`, `4x2x3` 등 곱이 24가 되는 모든 차원으로 데이터 유실 없이 유연하게 재구성될 수 있습니다.
2. **-1 와일드카드의 실무적 유용성:** 배치(Batch) 크기나 타임스텝 길이를 고정하고 샘플 수를 가변적으로 다룰 때 머신러닝 파이프라인에서 필수적으로 사용됩니다.


## 문제 6. flatten 평탄화와 불리언 마스킹(Boolean Masking) [고속 필터링]

2차원 시험 성적 데이터가 있습니다:
```python
exam_scores = np.array([
    [78, 85, 92],
    [54, 60, 88],
    [95, 42, 70]
])
```
요구사항:
1. `flatten()`을 사용하여 2차원 성적을 1차원 배열 `flat_scores`로 평탄화하세요.
2. 불리언 마스킹(`flat_scores >= 80`)을 생성하여 80점 이상인 우수 점수만 1차원 배열로 필터링하세요.
3. 80점 이상인 학생 점수들의 평균을 소수점 첫째 자리까지 구하세요.


````python
# 문제 6 풀이
exam_scores = np.array([
    [78, 85, 92],
    [54, 60, 88],
    [95, 42, 70]
])

# 1. 평탄화
flat_scores = exam_scores.flatten()
print("1차원 평탄화 결과:", flat_scores)

# 2. 불리언 마스킹 필터링
mask_80 = flat_scores >= 80
print("불리언 마스크 (80점 이상):", mask_80)
high_scores = flat_scores[mask_80]
print("80점 이상 점수 목록:", high_scores)

# 3. 우수 성적 평균
high_avg = high_scores.mean()
print(f"우수 성적 평균: {high_avg:.1f}점")
````

**결과 해석 / 확인한 점:**  
1. **불리언 마스킹의 메커니즘:** 조건식 `flat_scores >= 80`은 각 원소별로 참/거짓 배열(`[False, True, True, ...]`)을 반환하며, 이를 배열 색인에 전달하면 `True`인 위치의 값만 순차 추출됩니다.
2. **연산 간결성:** 반복문과 if문 없이 9개의 성적 중 85, 92, 88, 95 총 4건의 우수 성적이 추출되어 평균 90.0점이 정확히 계산되었습니다.
