# 7. 제곱합 분해와 결정계수

절편을 포함해 학습한 최소제곱 회귀의 훈련 표본에서는 전체 제곱합 SST가 회귀 제곱합 SSR과 오차 제곱합 SSE의 합입니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X = np.arange(1, 6).reshape(-1, 1)
y = np.array([1.5, 1.7, 3.2, 4.0, 5.1])
pred = LinearRegression().fit(X, y).predict(X)
```

## 문제 1. 제곱합 계산

SST, SSR, SSE를 직접 계산하고 제곱합 분해가 성립하는지 수치 오차를 허용해 확인하세요.

```python
mean = y.mean()
sst = np.sum((y - mean) ** 2)
ssr = np.sum((pred - mean) ** 2)
sse = np.sum((y - pred) ** 2)
print(sst, ssr, sse)
print(np.isclose(sst, ssr + sse))
```

**결과 해석**

절편을 포함한 최소제곱의 훈련 적합값에서는 잔차와 중심화한 적합값이 직교합니다.

## 문제 2. 결정계수 직접 계산

SSE와 SST로 R²를 구해 r2_score의 결과와 비교하세요.

```python
sse = np.sum((y - pred) ** 2)
sst = np.sum((y - y.mean()) ** 2)
manual_r2 = 1 - sse / sst
print(manual_r2, r2_score(y, pred), np.isclose(manual_r2, r2_score(y, pred)))
```

**결과 해석**

평균으로 예측할 때보다 제곱 오차가 클 경우 테스트 R²는 음수가 될 수 있습니다.
