# 16. 실루엣 스코어 군집 평가 (Silhouette Score)

군집 내부 응집도와 군집 간 분리도를 함께 측정하는 `silhouette_score`를 `k=2~10` 범위에서 계산해 최적 군집 수를 정량적으로 판별합니다.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=500, centers=4, cluster_std=0.60, random_state=0)
range_n_clusters = list(range(2, 11))
```

## 문제 1. k=2~10 평균 실루엣 스코어 계산

`range_n_clusters`에 대해 `KMeans(n_clusters=k, n_init=10, random_state=10)`로 군집화하고 `silhouette_score`를 계산해 출력하세요.

```python
silhouette_avg_scores = []
for k in range_n_clusters:
    km = KMeans(n_clusters=k, n_init=10, random_state=10)
    labels = km.fit_predict(X)
    score = silhouette_score(X, labels)
    silhouette_avg_scores.append(score)
    print(f'k={k}: silhouette_score={score:.3f}')
```

**결과 해석**

실루엣 스코어는 -1에서 1 사이의 값을 가지며 1에 가까울수록 응집도와 분리도가 모두 높습니다. 이 데이터에서는 `k=4`일 때 가장 높은 점수가 나타납니다.

## 문제 2. 실루엣 스코어 추세 시각화 및 최적 k 출력

`range_n_clusters`별 실루엣 스코어를 선 그래프로 그리고, 점수가 최대가 되는 최적의 `k`를 출력하세요.

```python
silhouette_avg_scores = []
for k in range_n_clusters:
    km = KMeans(n_clusters=k, n_init=10, random_state=10)
    labels = km.fit_predict(X)
    silhouette_avg_scores.append(silhouette_score(X, labels))

best_k = range_n_clusters[silhouette_avg_scores.index(max(silhouette_avg_scores))]
print(f'최적 군집 수: k={best_k} (score={max(silhouette_avg_scores):.3f})')

plt.figure(figsize=(8, 4))
plt.plot(range_n_clusters, silhouette_avg_scores, marker='o')
plt.title('Silhouette Scores for Various Numbers of Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Average Silhouette Score')
plt.show()
```

**결과 해석**

`k=4`에서 평균 실루엣 스코어가 정점을 찍고 `k=5` 이상부터는 군집이 과도하게 쪼개지면서 점수가 하락합니다.
