이 노트북은 `Numpy_2_Vectorized_Operations_Practice.ipynb`의 모든 연습문제에 대한 **표준 모범 코드와 결과 해석**을 제공합니다.


```python
# [실습 환경 준비]
# 이 셀은 실습을 위한 수치 데이터를 생성합니다. 셀을 먼저 실행해주세요.
import numpy as np

# 학생 5명의 4개 과목 시험 점수 행렬 (5행 4열: 국어, 영어, 수학, 과학)
exam_matrix = np.array([
    [85, 90, 95, 88],
    [70, 65, 80, 75],
    [95, 92, 98, 96],
    [60, 55, 62, 58],
    [80, 85, 82, 84]
])

# 상품 3개의 가격 및 판매 수량
unit_prices = np.array([15000, 28000, 45000]) # 1차원 벡터
sales_quantities = np.array([
    [2, 1, 0], # 월요일
    [3, 2, 1], # 화요일
    [1, 0, 2], # 수요일
    [4, 3, 2]  # 목요일
])
```

## 문제 1. 벡터화 연산을 통한 일일 매출액 산출 [벡터화 기초]

setup 코드의 월요일 판매 수량은 `monday_qty = np.array([2, 1, 0])`이고, 상품 단가는 `unit_prices = np.array([15000, 28000, 45000])`입니다.
1. `monday_qty`와 `unit_prices`의 요소별 곱셈(`*`)을 수행하여 상품별 발생 매출을 구하세요.
2. `np.sum()`을 사용하여 월요일의 총 매출액을 계산하고 출력하세요.
3. 반복문(`for`)을 전혀 사용하지 않고 단 두 줄로 계산을 완료하세요.


```python
# 문제 1 풀이
monday_qty = sales_quantities[0] # [2, 1, 0]

# 1. 요소별 곱셈 (Element-wise multiplication)
item_revenues = monday_qty * unit_prices
print("상품별 발생 매출:", item_revenues)

# 2. 총 매출 합산
total_monday_revenue = np.sum(item_revenues)
print(f"월요일 총 매출액: {total_monday_revenue:,}원")
```

**결과 해석 / 확인한 점:**  
1. **벡터화의 직관성:** `monday_qty * unit_prices`는 동일한 길이(3)를 갖는 두 벡터의 각 원소를 짝지어 2*15000=30000, 1*28000=28000, 0*45000=0으로 한 번에 계산합니다.
2. **실무 효율성:** 월요일 총 매출 58,000원이 반복문 없이 단일 수식으로 정확히 산출되었습니다.


## 문제 2. 브로드캐스팅(Broadcasting)을 활용한 과목별 평균 편차(Deviation) 계산 [브로드캐스팅]

setup의 `exam_matrix`는 5명 학생의 4과목 점수(5행 4열)입니다.
1. `axis=0`을 지정하여 4개 과목 각각의 평균 점수 벡터(`subject_means`, shape: `(4,)`)를 구하세요.
2. `exam_matrix - subject_means`를 수행하여 각 학생의 과목별 평균 대비 편차(Deviation) 행렬을 구하세요.
3. 브로드캐스팅이 어떻게 일어났는지 두 배열의 shape를 비교하여 출력하세요.


```python
# 문제 2 풀이
# 1. 과목별(열별) 평균 계산 (axis=0)
subject_means = exam_matrix.mean(axis=0)
print("과목별 평균 점수:", subject_means)
print("평균 벡터 shape:", subject_means.shape)

# 2. 브로드캐스팅을 통한 편차 행렬 산출
deviations = exam_matrix - subject_means
print("\n과목별 평균 대비 편차 행렬 (exam - mean):\n", deviations.round(1))

# 3. Shape 비교
print(f"\n원본 행렬 shape: {exam_matrix.shape} vs 평균 벡터 shape: {subject_means.shape}")
print("NumPy가 1차원 (4,) 벡터를 (5, 4) 형태로 가상 확장하여 차감했습니다.")
```

**결과 해석 / 확인한 점:**  
1. **브로드캐스팅 규칙:** 형상이 `(5, 4)`인 2차원 배열과 `(4,)`인 1차원 배열 연산 시, 마지막 축의 크기(4)가 일치하므로 행 방향으로 5번 자동 복제 확장되어 차원 맞춤이 일어납니다.
2. **데이터 중심화(Centering):** 머신러닝 주성분 분석(PCA) 등에서 피처의 평균을 0으로 맞추는 Mean-Centering 전처리가 브로드캐스팅 덕분에 단 한 줄로 수행됩니다.


## 문제 3. 축(axis) 지정 통계 집계: 학생별 총점과 과목별 표준편차 [축 집계]

`exam_matrix`를 활용하여 다음 두 가지 지표를 산출하세요:
1. 각 학생별 총점(`student_totals`, `axis=1`)과 평균(`student_averages`, `axis=1`)을 구하고 최고 득점 학생의 인덱스를 찾으세요 (`np.argmax()` 활용).
2. 과목별 점수의 표준편차(`subject_stds`, `axis=0`)를 소수점 둘째 자리까지 구하여 가장 성적 편차가 큰 과목의 인덱스를 찾으세요.


```python
# 문제 3 풀이
# 1. 학생별(행별, axis=1) 집계
student_totals = exam_matrix.sum(axis=1)
student_averages = exam_matrix.mean(axis=1)
top_student_idx = np.argmax(student_totals)

print("학생별 총점:", student_totals)
print("학생별 평균:", student_averages)
print(f"수석 학생: {top_student_idx}번 학생 (총점: {student_totals[top_student_idx]}점, 평균: {student_averages[top_student_idx]:.1f}점)")

# 2. 과목별(열별, axis=0) 표준편차
subject_stds = exam_matrix.std(axis=0).round(2)
most_spread_subject_idx = np.argmax(subject_stds)

print("\n과목별 표준편차 (국, 영, 수, 과):", subject_stds)
print(f"가장 편차가 큰 과목 인덱스: {most_spread_subject_idx}번 과목 (표준편차: {subject_stds[most_spread_subject_idx]})")
```

**결과 해석 / 확인한 점:**  
1. **축(axis)의 방향성:** `axis=1`은 행(가로줄)을 따라 압축하여 각 학생의 합계를 내고, `axis=0`은 열(세로줄)을 따라 압축하여 과목별 산포도를 측정합니다.
2. **최댓값 색인(argmax):** 수석 학생은 2번 학생(총점 381점, 평균 95.2점)이며, 성적 편차가 가장 큰 과목은 1번 과목(영어, 표준편차 14.28점)으로 변별력이 가장 높았음을 확인할 수 있습니다.


## 문제 4. 2차원 불리언 마스킹과 통계치 대체 [조건부 필터링]

`exam_matrix`에서 과락(Failing) 점수 기준을 `70점 미만`으로 정의합니다.
1. `exam_matrix < 70` 불리언 마스크를 생성하여 과락 점수가 총 몇 개인지 카운트하세요 (`np.sum()` 활용).
2. 70점 미만인 모든 과락 점수들을 추출하여 1차원 배열로 출력하세요.
3. 전체 점수 중 70점 미만을 제외한 '통과 점수(Passing Scores)'들의 평균을 계산하세요.


```python
# 문제 4 풀이
# 1. 불리언 마스크 및 과락 개수
fail_mask = exam_matrix < 70
fail_count = np.sum(fail_mask)
print("과락 점수 여부 마스크:\n", fail_mask)
print(f"총 과락 점수 개수: {fail_count}개")

# 2. 과락 점수 목록 추출
fail_scores = exam_matrix[fail_mask]
print("과락 점수 목록:", fail_scores)

# 3. 70점 이상 통과 점수 평균
pass_mask = exam_matrix >= 70
pass_scores = exam_matrix[pass_mask]
print(f"통과 점수들의 평균: {pass_scores.mean():.2f}점")
```

**결과 해석 / 확인한 점:**  
1. **불리언 배열의 합산(`sum`):** 파이썬에서 `True`는 숫자 1, `False`는 0으로 취급되므로, 불리언 마스크에 `np.sum()`을 취하면 조건을 만족하는 원소의 개수를 O(N)으로 즉시 구할 수 있습니다.
2. **필터링 결과:** 65점, 60점, 55점, 62점, 58점 총 5건의 과락 점수가 추출되었으며, 이를 제외한 정상 통과 점수 평균은 87.93점으로 계산되었습니다.


## 문제 5. np.where를 활용한 이상치/결손치 클리핑 및 등급 부여 [데이터 변환]

`np.where(condition, x, y)`는 조건에 따라 배열의 값을 교체하는 고속 삼항 연산자입니다.
1. `exam_matrix`에서 90점 이상인 점수는 `"PASS"`, 90점 미만인 점수는 `"NON-PASS"`로 변환한 문자열 행렬 `pass_matrix`를 생성하세요.
2. 수치 안정화를 위해, `exam_matrix`에서 65점 미만인 점수는 모두 최저 기준 점수인 `65`점으로 상향 보정(클리핑)한 `clipped_exam` 행렬을 생성하세요.


```python
# 문제 5 풀이
# 1. 90점 기준 PASS / NON-PASS 레이블 행렬
pass_matrix = np.where(exam_matrix >= 90, "PASS", "NON-PASS")
print("=== 90점 기준 합격 여부 행렬 ===")
print(pass_matrix)

# 2. 65점 미만 클리핑 (보정)
clipped_exam = np.where(exam_matrix < 65, 65, exam_matrix)
print("\n=== 최저 65점 보정 후 성적 행렬 ===")
print(clipped_exam)
```

**결과 해석 / 확인한 점:**  
1. **벡터화된 조건부 치환:** `np.where`는 파이썬의 `if-else` 분기 처리를 C 수준에서 병렬 수행하므로 데이터프레임의 파생변수 생성이나 결측 대체 시 최고의 속도를 자랑합니다.
2. **보정 검증:** 3번째 학생의 60점, 55점, 62점, 58점 등 65점 미만 점수들이 모두 65점으로 안전하게 클리핑(하한 보정)되었습니다.


## 문제 6. 행렬 곱(Matrix Multiplication: @)을 통한 요일별 총 매출 산출 [선형대수 응용]

setup 코드에 정의된:
- 요일별 판매량: `sales_quantities` (4행 3열: 월, 화, 수, 목 4일간 3개 상품의 판매 수량)
- 상품 단가: `unit_prices` (크기 3의 1차원 벡터: `[15000, 28000, 45000]`)

행렬 곱 연산자(`@` 또는 `np.dot`)를 사용하여:
1. `sales_quantities @ unit_prices`를 수행하여 월, 화, 수, 목 4일간의 일일 총 매출액 벡터 `daily_revenues` (shape: `(4,)`)를 한 번에 계산하세요.
2. 요일명 `['월요일', '화요일', '수요일', '목요일']`과 매핑하여 일일 매출액을 출력하세요.
3. 4일간 발생한 전체 매출의 총합을 구하세요.


```python
# 문제 6 풀이
days = ["월요일", "화요일", "수요일", "목요일"]

# 1. 행렬 곱 (@ 연산자): (4, 3) @ (3,) -> (4,)
daily_revenues = sales_quantities @ unit_prices

# 2. 요일별 매출 출력
print("=== 요일별 총 매출액 집계 (행렬 곱 @) ===")
for d, rev in zip(days, daily_revenues):
    print(f"{d}: {rev:>10,}원")

# 3. 4일간 총 매출
total_period_revenue = np.sum(daily_revenues)
print(f"\n4일간 누적 총 매출액: {total_period_revenue:,}원")
```

**결과 해석 / 확인한 점:**  
1. **행렬 곱(내적)의 위력:** `(4, 3)` 형상의 수량 행렬과 `(3,)` 형상의 단가 벡터를 행렬 곱하면 각 요일(행)의 수량과 단가가 자동으로 곱해진 뒤 합산되어 `(4,)` 벡터가 도출됩니다.
2. **실무적 의의:** 머신러닝의 다중 선형 회귀 예측식 $\hat{Y} = XW + b$ 및 신경망 순전파(Forward Pass) 연산의 수학적 기본 원리를 완벽하게 실현했습니다.
