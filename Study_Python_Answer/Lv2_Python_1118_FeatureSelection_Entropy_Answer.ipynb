# 3. 엔트로피와 특성 중요도

분류 트리는 엔트로피를 줄이는 분할을 찾습니다. 불순도 기반 중요도는 해당 모델이 사용한 신호를 요약합니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## 문제 1. 엔트로피 트리의 중요도

엔트로피 기준 트리를 학습하고 이름을 붙인 중요도를 내림차순 정렬하세요.

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(criterion='entropy', random_state=42).fit(X_train, y_train)
importance = pd.Series(model.feature_importances_, index=iris.feature_names)
print(importance.sort_values(ascending=False))
```

**결과 해석**

분할에서 불순도 감소에 기여한 특성의 점수가 커집니다. 이 값만으로 원인과 결과를 판단할 수 없습니다.

## 문제 2. 선택한 특성으로 다시 학습

훈련 세트에서 구한 중요도로 상위 2개 특성을 골라 전체 특성 트리와 테스트 정확도를 비교하세요.

```python
from sklearn.tree import DecisionTreeClassifier
full = DecisionTreeClassifier(criterion='entropy', random_state=42).fit(X_train, y_train)
top = full.feature_importances_.argsort()[-2:]
selected = DecisionTreeClassifier(criterion='entropy', random_state=42)
selected.fit(X_train[:, top], y_train)
print([iris.feature_names[i] for i in top])
print(full.score(X_test, y_test), selected.score(X_test[:, top], y_test))
```

**결과 해석**

특성 선택에도 학습이 포함되므로 선택은 훈련 데이터에서 진행하고 평가는 따로 합니다.
