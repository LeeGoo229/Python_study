# 6. 다중 선형 회귀와 Iris 데이터

Iris의 꽃받침 길이·너비, 꽃잎 길이로 꽃잎 너비를 예측합니다. 꽃 종류는 이 회귀의 목표 변수가 아닙니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
iris = load_iris()
X, y = iris.data[:, :3], iris.data[:, 3]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## 문제 1. 계수와 성능 출력

다중 회귀의 계수를 입력 특성명과 연결하고 테스트 MSE와 R²를 출력하세요.

```python
model = LinearRegression().fit(X_train, y_train)
print(dict(zip(iris.feature_names[:3], model.coef_)))
print('intercept:', model.intercept_)
pred = model.predict(X_test)
print(mean_squared_error(y_test, pred), r2_score(y_test, pred))
```

**결과 해석**

각 계수는 다른 입력을 고정했을 때 한 입력이 한 단위 증가할 때의 예측 변화량입니다.

## 문제 2. 실측과 예측 비교

실측값과 예측값의 산점도에 일치 기준선을 추가하세요.

```python
import matplotlib.pyplot as plt
pred = LinearRegression().fit(X_train, y_train).predict(X_test)
low, high = y_test.min(), y_test.max()
plt.scatter(y_test, pred)
plt.plot([low, high], [low, high], linestyle='--')
plt.xlabel('Actual petal width (cm)')
plt.ylabel('Predicted petal width (cm)')
plt.show()
```

**결과 해석**

점이 기준선보다 위에 있으면 관측값보다 크게 예측한 경우입니다.
