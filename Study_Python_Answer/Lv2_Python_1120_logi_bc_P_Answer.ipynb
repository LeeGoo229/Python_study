# 4. 로지스틱 회귀 유방암 실습 (Logistic Regression Breast Cancer)

유방암 데이터셋의 타깃 비율을 확인하고 표준화 + 로지스틱 회귀 모델을 학습·평가합니다.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

data = load_breast_cancer()
X = data.data
y = data.target
df = pd.DataFrame(X, columns=data.feature_names)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

## 문제 1. 데이터 탐색(EDA) 및 타깃 클래스 비율 확인

`df.shape`, 결측치 합계, 그리고 타깃 `y`(0: 악성, 1: 양성)의 클래스별 개수와 비율을 출력하세요.

```python
print('Shape:', df.shape, 'Missing values:', df.isnull().sum().sum())
s_y = pd.Series(y, name='target')
print('Counts:\n', s_y.value_counts())
print('Ratios:\n', s_y.value_counts(normalize=True).round(4))
```

**결과 해석**

총 569개 샘플·30개 수치형 특성에 결측치는 없으며, 양성(1) 약 62.7%·악성(0) 약 37.3%로 구성되어 있습니다.

## 문제 2. StandardScaler + LogisticRegression 학습 및 평가

훈련 세트에만 `StandardScaler`를 `fit_transform`하고 테스트 세트를 `transform`한 뒤 `LogisticRegression(max_iter=1000, random_state=42)`의 정확도, 혼동 행렬, 분류 리포트를 출력하세요.

```python
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)

clf = LogisticRegression(max_iter=1000, random_state=42)
clf.fit(X_train_sc, y_train)
y_pred = clf.predict(X_test_sc)

print(f'Accuracy: {accuracy_score(y_test, y_pred):.4f}')
print('Confusion Matrix:\n', confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred, target_names=data.target_names, digits=4))
```

**결과 해석**

30개 피처의 스케일을 표준화한 뒤 로지스틱 회귀를 학습하면 악성/양성 모두에서 98% 내외의 높은 정밀도와 재현율을 달성합니다.
