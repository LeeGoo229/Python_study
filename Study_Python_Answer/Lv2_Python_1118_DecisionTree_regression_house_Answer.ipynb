# 4. 의사결정나무와 랜덤 포레스트 회귀

회귀 트리는 자식 노드의 가중 제곱 오차가 낮아지도록 분할합니다. 랜덤 포레스트는 여러 트리의 예측을 평균합니다. 제공된 소규모 CSV의 점수는 분할에 민감합니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
from pathlib import Path
import pandas as pd

data_dir = DATA_DIR if 'DATA_DIR' in globals() else Path('data')
if not (data_dir / 'house.price.csv').is_file():
    data_dir = Path('Study_Python') / '[7] Machine_Learning' / 'data'
if not (data_dir / 'house.price.csv').is_file():
    data_dir = Path('..') / 'Study_Python' / '[7] Machine_Learning' / 'data'
file_path = data_dir / 'house.price.csv'
df = pd.read_csv(file_path)
from sklearn.model_selection import train_test_split

X = df[['year', 'size']]
y = df['price']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

## 문제 1. 후보 분할의 가중 오차

50, 60, 90, 100점을 앞뒤 두 그룹으로 나눴을 때 분할 전 MSE와 분할 후 가중 MSE를 구하세요.

```python
import numpy as np
scores = np.array([50, 60, 90, 100])
groups = np.split(scores, [2])
parent_mse = np.mean((scores - scores.mean()) ** 2)
child_mse = sum(np.sum((g - g.mean()) ** 2) for g in groups) / scores.size
print(parent_mse, child_mse)
```

**결과 해석**

분할 전 425에서 분할 후 25로 감소합니다. 그룹 크기는 가중치에 반영됩니다.

## 문제 2. 주택 가격 트리와 포레스트 비교

같은 분할에서 깊이 3인 회귀 트리와 랜덤 포레스트를 학습해 테스트 MSE와 R²를 비교하세요.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.tree import DecisionTreeRegressor
models = [DecisionTreeRegressor(max_depth=3, random_state=1),
          RandomForestRegressor(max_depth=3, n_estimators=200, random_state=1)]
scores = []
for model in models:
    pred = model.fit(X_train, y_train).predict(X_test)
    scores.append((type(model).__name__, mean_squared_error(y_test, pred), r2_score(y_test, pred)))
print(scores)
```

**결과 해석**

적은 테스트 관측치로 얻은 성능 순위는 분할이 달라지면 바뀔 수 있습니다.
