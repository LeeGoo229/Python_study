# 19. 고객 세분화 군집 분석 (Mall Customers Clustering)

`Mall_Customers.csv` 데이터를 표준화(`StandardScaler`)한 뒤 엘보우 기법과 `KMeans(n_clusters=5)`로 고객 군집을 세분화하고 군집별 평균 특성과 소비 패턴을 시각화합니다.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

candidates = [
    Path('Study_Python/[7] Machine_Learning/data/Mall_Customers.csv'),
    Path('../[7] Machine_Learning/data/Mall_Customers.csv'),
    Path('data/Mall_Customers.csv'),
]
csv_path = next((p for p in candidates if p.is_file()), None)
df = pd.read_csv(csv_path).drop(columns=['CustomerID'])
features = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
scaled_data = StandardScaler().fit_transform(df[features])
```

## 문제 1. KMeans(n_clusters=5) 군집화 및 군집별 평균 요약

표준화된 `scaled_data`에 `KMeans(n_clusters=5, n_init=10, random_state=42)`를 적용해 `df["Cluster_ID"]`를 생성하고 군집별 수치형 평균 요약표를 출력하세요.

```python
kmeans = KMeans(n_clusters=5, n_init=10, random_state=42)
df['Cluster_ID'] = kmeans.fit_predict(scaled_data)
print(df.select_dtypes(include=['number']).groupby('Cluster_ID').mean().round(2))
```

**결과 해석**

연령·연간 소득·소비 점수를 표준화해 5개 군집으로 나누면 고소득·고소비, 고소득·저소비 등 명확한 고객 세그먼트를 도출할 수 있습니다.

## 문제 2. 연간 소득과 소비 점수 군집 시각화

`sns.scatterplot`으로 `Annual Income (k$)`(x축)과 `Spending Score (1-100)`(y축)을 `Cluster_ID` 색상별로 시각화하세요.

```python
kmeans = KMeans(n_clusters=5, n_init=10, random_state=42)
df['Cluster_ID'] = kmeans.fit_predict(scaled_data)

plt.figure(figsize=(8, 5))
sns.scatterplot(x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster_ID', data=df, palette='viridis', s=80)
plt.title('Customer Segments: Income vs Spending Score')
plt.show()
```

**결과 해석**

소득 대비 소비 점수 분포를 시각화하면 마케팅 타깃이 되는 핵심 고객군(고소득·고소비 및 잠재 고소비군)을 직관적으로 식별할 수 있습니다.
