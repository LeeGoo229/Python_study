# 2. K-최근접 이웃 분류

KNN은 가까운 훈련 표본의 레이블을 이용해 새 표본의 범주를 예측합니다. 이웃 수, 거리 척도, 스케일링이 결과에 영향을 줍니다.

각 문제는 독립적으로 실행할 수 있습니다.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## 문제 1. 거리에 따른 이웃 투표

표준화와 거리 가중 KNN을 연결해 테스트 정확도와 분류 보고서를 출력하세요.

```python
from sklearn.metrics import classification_report
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5, weights='distance'))
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
print(classification_report(y_test, model.predict(X_test), zero_division=0))
```

**결과 해석**

거리 가중 투표는 가까운 표본에 더 큰 비중을 부여합니다. 분류 보고서로 범주별 정밀도와 재현율을 확인합니다.

## 문제 2. 이웃 수 및 거리 비교

훈련 세트의 교차검증에서 k=3, 9와 맨해튼·유클리드 거리 조합의 평균 정확도를 비교하세요.

```python
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
results = []
for k in (3, 9):
    for p in (1, 2):
        model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k, p=p))
        results.append((k, p, cross_val_score(model, X_train, y_train, cv=5).mean()))
print(results)
```

**결과 해석**

훈련 세트 안에서 거리와 k를 비교하면 테스트 세트를 최종 평가에 남겨둘 수 있습니다.
